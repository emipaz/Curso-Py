# 🎯 Plotly: Gráficos Interactivos y Dashboards

Bienvenidos a Plotly, la librería para crear visualizaciones web interactivas, modernas y profesionales.

## 🤔 ¿Qué es Plotly?

**Plotly** es una librería de JavaScript basada en plotly.js que permite crear gráficos interactivos para la web.

### Comparación: Matplotlib vs Seaborn vs Plotly

| Característica | Matplotlib | Seaborn | Plotly |
|---|---|---|---|
| **Interactividad** | No | No | Sí (zoom, hover, click) |
| **Web-ready** | No | No | Sí (HTML) |
| **Curva aprendizaje** | Moderada | Fácil | Fácil |
| **Personalización** | Máxima | Media | Alta |
| **Rendimiento** | Excelente | Excelente | Bueno (muchos puntos) |
| **Uso ideal** | Publicaciones científicas | Análisis exploratorio | Dashboards, web |
| **Animaciones** | Con esfuerzo | No | Sí (nativas) |
| **3D** | Sí | No | Sí |

### Cuándo usar cada una:
- **Matplotlib**: Papers científicos, control total
- **Seaborn**: Análisis exploratorio rápido
- **Plotly**: Dashboards, presentaciones, exploradores web interactivos

## 📜 Historia

**Fundadores:** Jack Parmer y Chris Parmer  
**Año:** 2013  
**Motivación:** Crear una alternativa moderna e interactiva a herramientas de visualización existentes, pensada desde el principio para la web.

Plotly se convirtió en herramienta estándar para crear dashboards interactivos en Python.

```python
# !pip install plotly kaleido pandas numpy

import plotly.graph_objects as go
import plotly.express as px
import pandas as pd
import numpy as np

# Verificar versión
import plotly
print(f"Versión de Plotly: {plotly.__version__}")

# Configurar para mejor visualización
import warnings
warnings.filterwarnings('ignore')
```

## 📈 Gráficos Básicos con Plotly Express

**Plotly Express** es la interfaz de alto nivel (simplificada).

```python
# Cargar datos
tips = px.data.tips()
print(tips.head())
```

### 1. Scatter Plot (Gráfico de Dispersión)

```python
# Scatter plot simple
fig = px.scatter(tips, x='total_bill', y='tip', 
                 title='Scatter Plot: Total Bill vs Tip',
                 labels={'total_bill': 'Total Bill ($)', 'tip': 'Tip ($)'})
fig.show()
```

```python
# Scatter plot con colores y tamaños
fig = px.scatter(tips, x='total_bill', y='tip',
                 color='sex', size='party_size',
                 hover_data=['day', 'time'],
                 title='Scatter Plot Interactivo',
                 labels={'total_bill': 'Total Bill ($)', 'tip': 'Tip ($)'},
                 color_discrete_map={'Male': '#1f77b4', 'Female': '#ff7f0e'})
fig.update_traces(marker=dict(size=12, opacity=0.7))
fig.show()
```

### 2. Gráficos de Línea

```python
# Serie temporal
flights = px.data.flights()
fig = px.line(flights, x='month', y='passengers', color='year',
              title='Pasajeros por Mes (1949-1960)',
              labels={'passengers': 'Número de Pasajeros', 'month': 'Mes'})
fig.update_layout(hovermode='x unified')
fig.show()
```

### 3. Gráficos de Barras

```python
# Bar plot simple
fig = px.bar(tips, x='day', y='total_bill', color='sex',
             title='Total Bill por Día y Sexo',
             barmode='group',
             labels={'total_bill': 'Total Bill ($)', 'day': 'Día'})
fig.show()
```

```python
# Bar plot horizontal
conteo = tips.groupby('day').size().reset_index(name='count')
fig = px.bar(conteo, y='day', x='count', 
             title='Conteo de Visitas por Día',
             orientation='h',
             labels={'count': 'Número de Visitas', 'day': 'Día'})
fig.show()
```

### 4. Gráficos de Distribución

```python
# Histograma
fig = px.histogram(tips, x='total_bill', nbins=20,
                   title='Distribución del Total Bill',
                   labels={'total_bill': 'Total Bill ($)'},
                   marginal='box')
fig.show()
```

```python
# Box plot
fig = px.box(tips, x='day', y='total_bill', color='sex',
             title='Box Plot: Distribución por Día',
             labels={'total_bill': 'Total Bill ($)', 'day': 'Día'})
fig.show()
```

```python
# Violin plot
fig = px.violin(tips, x='day', y='total_bill', color='sex', box=True,
                title='Violin Plot con Box Plot',
                labels={'total_bill': 'Total Bill ($)', 'day': 'Día'})
fig.show()
```

### 5. Gráficos de Pastel

```python
# Pie chart
conteo_sexo = tips['sex'].value_counts().reset_index()
conteo_sexo.columns = ['sex', 'count']

fig = px.pie(conteo_sexo, values='count', names='sex',
             title='Distribución por Sexo',
             color_discrete_sequence=['#1f77b4', '#ff7f0e'])
fig.show()
```

### 6. Heatmaps

```python
# Matriz de correlación
iris = px.data.iris()
corr_matrix = iris.corr(numeric_only=True)

fig = px.imshow(corr_matrix, 
                 labels=dict(color="Correlación"),
                 title='Matriz de Correlación - Iris',
                 color_continuous_scale='RdBu_r',
                 zmin=-1, zmax=1)
fig.update_xaxes(side="bottom")
fig.show()
```

### 7. Gráficos 3D

```python
# Scatter 3D
fig = px.scatter_3d(iris, x='sepal_length', y='sepal_width', z='petal_length',
                    color='species',
                    title='Gráfico 3D - Iris Dataset',
                    labels={'species': 'Especie'})
fig.show()
```

```python
# Surface 3D
x = np.linspace(-5, 5, 50)
y = np.linspace(-5, 5, 50)
X, Y = np.meshgrid(x, y)
Z = np.sin(np.sqrt(X**2 + Y**2))

fig = go.Figure(data=[go.Surface(x=X, y=Y, z=Z, colorscale='Viridis')])
fig.update_layout(title='Superficie 3D Interactiva',
                  scene=dict(xaxis_title='X', yaxis_title='Y', zaxis_title='Z'))
fig.show()
```

### 8. Mapas de Burbujas y Sunburst

```python
# Bubble chart
fig = px.scatter(tips, x='total_bill', y='tip', size='party_size',
                 color='day', hover_data=['sex', 'time'],
                 title='Bubble Chart: Bill, Tip y Party Size',
                 labels={'total_bill': 'Total Bill ($)', 'tip': 'Tip ($)', 'party_size': 'Tamaño Grupo'})
fig.show()
```

```python
# Sunburst chart
fig = px.sunburst(tips, path=['sex', 'day', 'time'], values='total_bill',
                   title='Sunburst Chart: Jerarquía de Datos',
                   color='total_bill', color_continuous_scale='RdBu')
fig.show()
```

### 9. Facet Plots (Subgráficos)

```python
# Facet grid
fig = px.scatter(tips, x='total_bill', y='tip',
                 facet_col='sex', facet_row='time',
                 color='day', size='party_size',
                 title='Facet Plot: Análisis Multidimensional',
                 labels={'total_bill': 'Total Bill ($)', 'tip': 'Tip ($)'})
fig.show()
```

## 🎛️ Graph Objects: Control de Bajo Nivel

```python
# Crear gráfico con control total usando graph_objects
fig = go.Figure()

# Agregar trazos (traces)
x_data = np.linspace(0, 10, 100)

fig.add_trace(go.Scatter(
    x=x_data,
    y=np.sin(x_data),
    mode='lines+markers',
    name='sin(x)',
    line=dict(color='blue', width=2),
    marker=dict(size=5)
))

fig.add_trace(go.Scatter(
    x=x_data,
    y=np.cos(x_data),
    mode='lines+markers',
    name='cos(x)',
    line=dict(color='red', width=2, dash='dash'),
    marker=dict(size=5)
))

# Personalizar layout
fig.update_layout(
    title='Funciones Trigonométricas Interactivas',
    xaxis_title='X',
    yaxis_title='Y',
    hovermode='x unified',
    template='plotly_white',
    font=dict(size=12),
    height=600
)

fig.show()
```

### Gráfico con Anotaciones

```python
# Crear gráfico con anotaciones interactivas
fig = go.Figure()

# Datos
x_data = [1, 2, 3, 4, 5]
y_data = [10, 15, 13, 17, 20]

fig.add_trace(go.Scatter(
    x=x_data, y=y_data,
    mode='markers+lines',
    name='Datos',
    marker=dict(size=12, color='red'),
    line=dict(width=2)
))

# Agregar anotaciones
for i, (x, y) in enumerate(zip(x_data, y_data)):
    fig.add_annotation(
        x=x, y=y,
        text=f'Punto {i+1}<br>({x}, {y})',
        showarrow=True,
        arrowhead=2,
        ax=50,
        ay=-50
    )

fig.update_layout(
    title='Gráfico con Anotaciones Interactivas',
    xaxis_title='X',
    yaxis_title='Y',
    template='plotly_white',
    height=600
)

fig.show()
```

## 🎞️ Animaciones con Plotly

```python
# Animación de dispersión
gapminder = px.data.gapminder()

fig = px.scatter(gapminder, x="gdpPercap", y="lifeExp",
                 animation_frame="year", animation_group="country",
                 size="pop", color="continent",
                 hover_name="country", 
                 log_x=True, size_max=55,
                 range_x=[100, 100000], range_y=[25, 90],
                 title='Animación: PIB per Cápita vs Expectativa de Vida (1952-2007)',
                 labels={'gdpPercap': 'PIB per Cápita (log)', 'lifeExp': 'Expectativa de Vida (años)'})

fig.show()
```

```python
# Animación de barras
fig = px.bar(gapminder.query("continent == 'Europe'"),
             x="country", y="lifeExp",
             animation_frame="year",
             range_y=[30, 90],
             title='Esperanza de Vida en Europa (1952-2007)')

fig.show()
```

## 🎨 Widgets Interactivos con Plotly

```python
from ipywidgets import interact, Dropdown, IntSlider

@interact
def explorador_plotly(
    tipo_grafico=Dropdown(options=['scatter', 'line', 'bar', 'box'], value='scatter'),
    tamaño_figura=IntSlider(min=400, max=800, step=100, value=600)
):
    if tipo_grafico == 'scatter':
        fig = px.scatter(tips, x='total_bill', y='tip', color='sex',
                        title='Scatter Plot Interactivo',
                        height=tamaño_figura)
    elif tipo_grafico == 'line':
        fig = px.line(flights, x='month', y='passengers', color='year',
                     title='Line Plot Interactivo',
                     height=tamaño_figura)
    elif tipo_grafico == 'bar':
        fig = px.bar(tips, x='day', y='total_bill', color='sex',
                    title='Bar Plot Interactivo',
                    barmode='group',
                    height=tamaño_figura)
    else:  # box
        fig = px.box(tips, x='day', y='total_bill', color='sex',
                    title='Box Plot Interactivo',
                    height=tamaño_figura)
    
    fig.show()
```

## 📊 Dashboards con Plotly

```python
# Crear un dashboard con múltiples gráficos
from plotly.subplots import make_subplots

# Crear subplots
fig = make_subplots(
    rows=2, cols=2,
    subplot_titles=('Scatter', 'Histograma', 'Box Plot', 'Pie Chart'),
    specs=[
        [{'type': 'scatter'}, {'type': 'histogram'}],
        [{'type': 'box'}, {'type': 'pie'}]
    ]
)

# Scatter plot
fig.add_trace(
    go.Scatter(x=tips['total_bill'], y=tips['tip'], 
               mode='markers', name='Tips',
               marker=dict(size=8, color='blue', opacity=0.6)),
    row=1, col=1
)

# Histograma
fig.add_trace(
    go.Histogram(x=tips['total_bill'], name='Bill Distribution',
                nbinsx=20, marker=dict(color='green')),
    row=1, col=2
)

# Box plot
for day in tips['day'].unique():
    fig.add_trace(
        go.Box(y=tips[tips['day']==day]['total_bill'], name=day),
        row=2, col=1
    )

# Pie chart
conteo = tips['sex'].value_counts()
fig.add_trace(
    go.Pie(labels=conteo.index, values=conteo.values, name='Sex Distribution'),
    row=2, col=2
)

# Actualizar ejes
fig.update_xaxes(title_text='Total Bill ($)', row=1, col=1)
fig.update_yaxes(title_text='Tip ($)', row=1, col=1)
fig.update_xaxes(title_text='Total Bill ($)', row=1, col=2)
fig.update_yaxes(title_text='Total Bill ($)', row=2, col=1)

fig.update_layout(height=800, title_text='Dashboard Interactivo: Dataset Tips',
                 showlegend=True, template='plotly_white')

fig.show()
```
