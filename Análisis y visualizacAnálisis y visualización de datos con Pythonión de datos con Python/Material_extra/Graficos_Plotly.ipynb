# Ejemplos de Gráficos Dinámicos con Plotly

Este notebook muestra cómo crear gráficos interactivos y dinámicos usando Plotly en Python.

## 1. Importar Librerías
Importamos plotly y pandas para los ejemplos.

````python
import plotly.express as px
import plotly.graph_objects as go
import pandas as pd
import numpy as np
````

## 2. Gráfico de Línea Interactivo
Ideal para explorar tendencias y valores con el mouse.

````python
x = np.arange(0, 10, 0.1)
y = np.sin(x)
fig = px.line(x=x, y=y, labels={'x': 'Eje X', 'y': 'Seno(x)'}, title='Gráfico de Línea Interactivo')
fig.show()
````

## 3. Gráfico de Dispersión Interactivo
Permite explorar relaciones entre variables y ver detalles al pasar el mouse.

````python
np.random.seed(1)
df = pd.DataFrame({
    'x': np.random.rand(100),
    'y': np.random.rand(100),
    'categoria': np.random.choice(['A', 'B', 'C'], 100)
})
fig = px.scatter(df, x='x', y='y', color='categoria', title='Gráfico de Dispersión Interactivo',
                      labels={'x': 'Variable X', 'y': 'Variable Y'})
fig.show()
````

## 4. Histograma Interactivo
Para explorar la distribución de una variable numérica.

````python
datos = np.random.normal(0, 1, 500)
fig = px.histogram(datos, nbins=30, title='Histograma Interactivo', labels={'value': 'Valor'})
fig.show()
````

## 5. Gráfico de Barras Interactivo
Ideal para comparar categorías y ver detalles al interactuar.

````python
categorias = ['A', 'B', 'C', 'D']
valores = [23, 45, 12, 36]
fig = px.bar(x=categorias, y=valores, labels={'x': 'Categoría', 'y': 'Valor'},
                title='Gráfico de Barras Interactivo')
fig.show()
````

## 6. Gráfico de Pastel Interactivo
Para mostrar proporciones de un total de forma dinámica.

````python
labels = ['Manzanas', 'Bananas', 'Cerezas', 'Dátiles']
sizes = [30, 25, 25, 20]
fig = px.pie(names=labels, values=sizes, title='Gráfico de Pastel Interactivo')
fig.show()
````

## 7. Boxplot Interactivo
Para analizar la dispersión y outliers de una variable.

````python
df_box = pd.DataFrame({
    'grupo': np.random.choice(['G1', 'G2'], 200),
    'valor': np.random.normal(50, 15, 200)
})
fig = px.box(df_box, x='grupo', y='valor', color='grupo', title='Boxplot Interactivo')
fig.show()
````

## 8. Gráfico de Área Interactivo
Para mostrar evolución acumulada o proporciones en el tiempo.

````python
x = np.arange(0, 10, 0.1)
y1 = np.sin(x)
y2 = np.cos(x)
fig = go.Figure()
fig.add_trace(go.Scatter(x=x, y=y1, fill='tozeroy', name='Seno'))
fig.add_trace(go.Scatter(x=x, y=y2, fill='tozeroy', name='Coseno'))
fig.update_layout(title='Gráfico de Área Interactivo', xaxis_title='x', yaxis_title='Valor')
fig.show()
````

## 9. Gráfico de Mapas Interactivo

Plotly permite crear mapas para visualizar datos geográficos de manera dinámica.

````python
# Mapa de puntos geográficos
import plotly.express as px

df_map = pd.DataFrame({
    'ciudad': ['Buenos Aires', 'Madrid', 'París', 'Roma', 'Londres'],
    'lat': [-34.6037, 40.4168, 48.8566, 41.9028, 51.5074],
    'lon': [-58.3816, -3.7038, 2.3522, 12.4964, -0.1278],
    'valor': [100, 200, 150, 120, 180]
})

fig = px.scatter_mapbox(df_map, lat='lat', lon='lon', hover_name='ciudad', size='valor',
                        color='valor', color_continuous_scale='Viridis', zoom=1,
                        mapbox_style='carto-positron', title='Mapa Interactivo de Ciudades')
fig.show()
````

## 10. Gráfico de Series Temporales Interactivo

Ideal para analizar la evolución de variables a lo largo del tiempo.

````python
# Serie temporal interactiva
fechas = pd.date_range('2023-01-01', periods=100)
valores = np.random.normal(100, 10, 100).cumsum()
df_time = pd.DataFrame({'fecha': fechas, 'valor': valores})

fig = px.line(df_time, x='fecha', y='valor', title='Serie Temporal Interactiva', labels={'valor': 'Valor', 'fecha': 'Fecha'})
fig.update_xaxes(rangeslider_visible=True)
fig.show()
````

## 11. Gráfico de Burbujas Interactivo

Permite visualizar tres variables numéricas en un solo gráfico.

````python
# Gráfico de burbujas
np.random.seed(2)
df_bubble = pd.DataFrame({
    'x': np.random.rand(50),
    'y': np.random.rand(50),
    'tamaño': np.random.randint(10, 100, 50),
    'color': np.random.choice(['A', 'B', 'C'], 50)
})
fig = px.scatter(df_bubble, x='x', y='y', size='tamaño', color='color',
                 title='Gráfico de Burbujas Interactivo', labels={'x': 'Eje X', 'y': 'Eje Y'})
fig.show()
````

## 12. Gráfico de Violin Interactivo

Ideal para comparar la distribución de varias categorías.

````python
# Gráfico de violín
np.random.seed(3)
df_violin = pd.DataFrame({
    'grupo': np.random.choice(['A', 'B', 'C'], 300),
    'valor': np.random.normal(0, 1, 300)
})
fig = px.violin(df_violin, y='valor', x='grupo', color='grupo', box=True, points='all',
                title='Gráfico de Violin Interactivo')
fig.show()
````

## 13. Gráfico de Densidad 2D Interactivo

Permite visualizar la concentración de puntos en dos dimensiones.

````python
# Gráfico de densidad 2D
np.random.seed(4)
x = np.random.randn(500)
y = np.random.randn(500)
fig = px.density_heatmap(x=x, y=y, nbinsx=30, nbinsy=30, color_continuous_scale='Viridis',
                        title='Gráfico de Densidad 2D Interactivo', labels={'x': 'Eje X', 'y': 'Eje Y'})
fig.show()
````
