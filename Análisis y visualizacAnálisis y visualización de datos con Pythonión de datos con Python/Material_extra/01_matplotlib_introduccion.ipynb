## 🤔 ¿Qué es Matplotlib?

**Matplotlib** es una librería de Python de código abierto que permite crear visualizaciones estáticas, animadas e interactivas de alta calidad. Es una herramienta fundamental en ciencia de datos, análisis estadístico y visualización científica.

### Características principales:
- 📈 Crear gráficos 2D y 3D
- 🎨 Personalización completa (colores, estilos, fuentes)
- 📁 Exportar en múltiples formatos (PNG, PDF, SVG)
- 🔄 Compatible con NumPy y Pandas
- 💻 Funciona en scripts, notebooks y aplicaciones
- 🌍 Comunidad activa con abundante documentación

## 📜 Historia y Origen

## 🎯 Conceptos Fundamentales

Matplotlib funciona con una jerarquía de objetos:

```
Figure (la ventana general)
  └── Axes (el área donde se dibuja)
      ├── Líneas
      ├── Puntos
      ├── Barras
      └── Leyendas, etiquetas, etc.
```

### Los dos estilos de Matplotlib:

1. **Estilo MATLAB (pyplot)** - Más simple para gráficos rápidos
2. **Orientado a Objetos** - Más control y flexibilidad

## 📈 Tu Primer Gráfico

````python
# Estilo simple con pyplot
x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.figure(figsize=(10, 6))
plt.plot(x, y, linewidth=2, color='blue', label='sin(x)')
plt.title('Mi Primer Gráfico - Función Seno', fontsize=16, fontweight='bold')
plt.xlabel('X', fontsize=12)
plt.ylabel('Y = sin(X)', fontsize=12)
plt.legend(fontsize=10)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
````

## 🎨 Tipos de Gráficos Básicos

````python
# Crear una figura con múltiples subgráficos
fig, axes = plt.subplots(2, 3, figsize=(15, 10))
fig.suptitle('Gráficos Básicos en Matplotlib', fontsize=16, fontweight='bold')

# Datos de ejemplo
x = np.linspace(0, 10, 50)
y = np.sin(x)
categorias = ['A', 'B', 'C', 'D', 'E']
valores = [23, 45, 56, 78, 32]

# 1. Gráfico de línea
axes[0, 0].plot(x, y, 'b-', linewidth=2)
axes[0, 0].set_title('Gráfico de Línea')
axes[0, 0].set_xlabel('X')
axes[0, 0].set_ylabel('sin(X)')
axes[0, 0].grid(True, alpha=0.3)

# 2. Gráfico de barras
axes[0, 1].bar(categorias, valores, color='steelblue')
axes[0, 1].set_title('Gráfico de Barras')
axes[0, 1].set_ylabel('Valores')

# 3. Gráfico de dispersión
x_scatter = np.random.randn(100)
y_scatter = np.random.randn(100)
axes[0, 2].scatter(x_scatter, y_scatter, alpha=0.6, s=100)
axes[0, 2].set_title('Gráfico de Dispersión')
axes[0, 2].set_xlabel('X')
axes[0, 2].set_ylabel('Y')

# 4. Histograma
datos = np.random.normal(100, 15, 1000)
axes[1, 0].hist(datos, bins=30, color='green', alpha=0.7, edgecolor='black')
axes[1, 0].set_title('Histograma')
axes[1, 0].set_xlabel('Valores')
axes[1, 0].set_ylabel('Frecuencia')

# 5. Gráfico de pastel
axes[1, 1].pie(valores, labels=categorias, autopct='%1.1f%%', startangle=90)
axes[1, 1].set_title('Gráfico de Pastel')

# 6. Gráfico de área
x_area = np.linspace(0, 10, 100)
y1 = np.sin(x_area)
y2 = np.cos(x_area)
axes[1, 2].fill_between(x_area, y1, alpha=0.3, label='sin(x)')
axes[1, 2].fill_between(x_area, y2, alpha=0.3, label='cos(x)')
axes[1, 2].plot(x_area, y1, 'b-', linewidth=2)
axes[1, 2].plot(x_area, y2, 'r-', linewidth=2)
axes[1, 2].set_title('Gráfico de Área')
axes[1, 2].legend()
axes[1, 2].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
````

## 🎯 Personalización Avanzada

````python
# Crear un gráfico con personalización completa


fig, ax = plt.subplots(figsize=(12, 7))

# Datos
meses = ['Ene', 'Feb', 'Mar', 'Abr', 'May', 'Jun', 'Jul', 'Ago', 'Sep', 'Oct', 'Nov', 'Dic']
ventas_a = [10, 12, 15, 18, 22, 25, 23, 26, 28, 32, 35, 38]
ventas_b = [8, 10, 12, 14, 16, 18, 19, 20, 22, 24, 26, 28]

x_pos = np.arange(len(meses))
ancho = 0.35

# Gráficos de barras
barras1 = ax.bar(x_pos - ancho/2, ventas_a, ancho, label='Producto A', 
                  color='#2E86AB', alpha=0.8, edgecolor='black', linewidth=1.2)
barras2 = ax.bar(x_pos + ancho/2, ventas_b, ancho, label='Producto B', 
                  color='#A23B72', alpha=0.8, edgecolor='black', linewidth=1.2)

# Añadir valores en las barras
for barra in barras1:
    altura = barra.get_height()
    ax.text(barra.get_x() + barra.get_width()/2., altura,
            f'{int(altura)}', ha='center', va='bottom', fontsize=9)

for barra in barras2:
    altura = barra.get_height()
    ax.text(barra.get_x() + barra.get_width()/2., altura,
            f'{int(altura)}', ha='center', va='bottom', fontsize=9)

# Personalizar el gráfico
ax.set_xlabel('Mes', fontsize=12, fontweight='bold')
ax.set_ylabel('Ventas (en miles)', fontsize=12, fontweight='bold')
ax.set_title('Comparación de Ventas Mensuales', fontsize=14, fontweight='bold', pad=20)
ax.set_xticks(x_pos)
ax.set_xticklabels(meses)
ax.legend(fontsize=11, loc='upper left')
ax.grid(True, axis='y', alpha=0.3, linestyle='--')
ax.set_facecolor('#F5F5F5')

plt.tight_layout()
plt.show()
````

## 🎛️ Widgets Interactivos con Matplotlib

Matplotlib puede ser interactivo usando ipywidgets. Vamos a crear gráficos dinámicos que respondan a cambios de parámetros.

````python
# Si no tienes ipywidgets instalado:
# !pip install ipywidgets
# !pip install jupyterlab_widget

from ipywidgets import interact, FloatSlider, IntSlider, Dropdown
import matplotlib.pyplot as plt
import numpy as np



print("Widgets disponibles para interactividad en Matplotlib")
````

````python
# Widget 1: Explorador de Funciones Trigonométricas
@interact(
    funcion=Dropdown(options=['sin', 'cos', 'exp'], value='sin'),
    amplitud=FloatSlider(min=0.1, max=3, step=0.1, value=1),
    frecuencia=FloatSlider(min=0.5, max=3, step=0.1, value=1),
    fase=FloatSlider(min=0, max=2*np.pi, step=0.1, value=0)
)
def explorador_trigonometrico(funcion='sin', amplitud=1, frecuencia=1, fase=0):
    x = np.linspace(-2*np.pi, 2*np.pi, 1000)
    
    # Seleccionar función
    if funcion == 'sin':
        y = amplitud * np.sin(frecuencia * x + fase)
        titulo = f'y = {amplitud:.2f} × sin({frecuencia:.2f}x + {fase:.2f})'
    elif funcion == 'cos':
        y = amplitud * np.cos(frecuencia * x + fase)
        titulo = f'y = {amplitud:.2f} × cos({frecuencia:.2f}x + {fase:.2f})'
    else:  # exponencial amortiguada
        y = amplitud * np.exp(-np.abs(x)/3) * np.sin(frecuencia * x + fase)
        titulo = f'y = {amplitud:.2f} × exp(-|x|/3) × sin({frecuencia:.2f}x + {fase:.2f})'
    
    plt.figure(figsize=(12, 6))
    plt.plot(x, y, linewidth=2.5, color='#FF6B6B')
    plt.title(titulo, fontsize=14, fontweight='bold')
    plt.xlabel('X', fontsize=12)
    plt.ylabel('Y', fontsize=12)
    plt.grid(True, alpha=0.3)
    plt.ylim(-4, 4)
    plt.axhline(y=0, color='k', linewidth=0.5)
    plt.axvline(x=0, color='k', linewidth=0.5)
    plt.tight_layout()
    plt.show()
````

````python
# Widget 2: Generador de Datos Aleatorios con Visualización
@interact
def generador_distribucion(
    num_puntos=IntSlider(min=10, max=5000, step=100, value=1000),
    media=FloatSlider(min=-50, max=50, step=5, value=0),
    desv_estandar=FloatSlider(min=1, max=50, step=1, value=15),
    num_bins=IntSlider(min=10, max=100, step=5, value=30)
):
    
    datos = np.random.normal(media, desv_estandar, num_puntos)
    
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
    
    # Histograma
    ax1.hist(datos, bins=num_bins, color='#4ECDC4', alpha=0.7, edgecolor='black')
    ax1.set_title('Distribución Normal', fontsize=12, fontweight='bold')
    ax1.set_xlabel('Valores')
    ax1.set_ylabel('Frecuencia')
    ax1.grid(True, alpha=0.3)
    
    # Box plot
    ax2.boxplot(datos, orientation="vertical")
    ax2.set_title('Box Plot', fontsize=12, fontweight='bold')
    ax2.set_ylabel('Valores')
    ax2.grid(True, alpha=0.3)
    
    # Estadísticas
    stats_text = f"Media: {datos.mean():.2f}\nDesv Est: {datos.std():.2f}\nMin: {datos.min():.2f}\nMax: {datos.max():.2f}"
    ax2.text(1.15, np.median(datos), stats_text, fontsize=10, 
             bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))
    
    plt.tight_layout()
    plt.show()
````

````python
# Widget 3: Explorador de Colores y Estilos
@interact
def explorador_estilos(
    color=Dropdown(options=['blue', 'red', 'green', 'purple', 'orange', 'cyan'], value='blue'),
    estilo_linea=Dropdown(options=['-', '--', '-.', ':'], value='-'),
    ancho_linea=FloatSlider(min=0.5, max=5, step=0.5, value=2)
):
    
    x = np.linspace(0, 10, 1000)
    y = np.sin(x) * np.exp(-x/10)
    
    plt.figure(figsize=(12, 6))
    plt.plot(x, y, color=color, linestyle=estilo_linea, linewidth=ancho_linea, label='Función')
    plt.title(f'Color: {color} | Estilo: {estilo_linea} | Ancho: {ancho_linea}', 
              fontsize=14, fontweight='bold')
    plt.xlabel('X')
    plt.ylabel('Y')
    plt.legend(fontsize=12)
    plt.grid(True, alpha=0.3)
    plt.show()
````

## 🌟 Las Infinitas Posibilidades de Matplotlib

Matplotlib es increíblemente versátil. Aquí hay una muestra de lo que puedes crear:

### 1️⃣ Gráficos en 3D

````python
from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure(figsize=(12, 5))

# Gráfico 3D - Superficie
ax1 = fig.add_subplot(121, projection='3d')
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(x, y)
Z = np.sin(np.sqrt(X**2 + Y**2))

surf = ax1.plot_surface(X, Y, Z, cmap='viridis', alpha=0.8)
ax1.set_title('Superficie 3D')
ax1.set_xlabel('X')
ax1.set_ylabel('Y')
ax1.set_zlabel('Z')
fig.colorbar(surf, ax=ax1, pad=0.1)

# Gráfico 3D - Dispersión
ax2 = fig.add_subplot(122, projection='3d')
n_puntos = 200
xs = np.random.randn(n_puntos)
ys = np.random.randn(n_puntos)
zs = np.random.randn(n_puntos)
colors = np.sqrt(xs**2 + ys**2 + zs**2)

scatter = ax2.scatter(xs, ys, zs, c=colors, cmap='plasma', s=50, alpha=0.6)
ax2.set_title('Dispersión 3D')
ax2.set_xlabel('X')
ax2.set_ylabel('Y')
ax2.set_zlabel('Z')

plt.tight_layout()
plt.show()
````

### 2️⃣ Heatmaps y Mapas de Calor

````python
# Crear una matriz de datos
datos_matriz = np.random.randn(10, 10) * 10 + 50

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Heatmap con valores
im1 = ax1.imshow(datos_matriz, cmap='RdYlGn', aspect='auto')
ax1.set_title('Heatmap: RdYlGn', fontsize=12, fontweight='bold')
ax1.set_xlabel('Columnas')
ax1.set_ylabel('Filas')
cbar1 = plt.colorbar(im1, ax=ax1)
cbar1.set_label('Intensidad')

# Anotaciones
for i in range(datos_matriz.shape[0]):
    for j in range(datos_matriz.shape[1]):
        text = ax1.text(j, i, f'{datos_matriz[i, j]:.0f}',
                        ha="center", va="center", color="black", fontsize=8)

# Heatmap con otro colormap
im2 = ax2.imshow(datos_matriz, cmap='coolwarm', aspect='auto')
ax2.set_title('Heatmap: Coolwarm', fontsize=12, fontweight='bold')
ax2.set_xlabel('Columnas')
ax2.set_ylabel('Filas')
cbar2 = plt.colorbar(im2, ax=ax2)
cbar2.set_label('Intensidad')

plt.tight_layout()
plt.show()
````

### 3️⃣ Gráficos de Contorno y Campos Vectoriales

````python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

# Datos
x = np.linspace(-3, 3, 100)
y = np.linspace(-3, 3, 100)
X, Y = np.meshgrid(x, y)
Z = np.sin(np.sqrt(X**2 + Y**2)) * np.exp(-0.1*(X**2 + Y**2))

# Gráfico de contorno
contour = ax1.contourf(X, Y, Z, levels=20, cmap='viridis')
contour_lines = ax1.contour(X, Y, Z, levels=10, colors='white', alpha=0.3, linewidths=0.5)
ax1.clabel(contour_lines, inline=True, fontsize=8)
ax1.set_title('Gráfico de Contorno Lleno', fontsize=12, fontweight='bold')
ax1.set_xlabel('X')
ax1.set_ylabel('Y')
plt.colorbar(contour, ax=ax1)

# Campo vectorial
x_vec = np.linspace(-3, 3, 20)
y_vec = np.linspace(-3, 3, 20)
X_vec, Y_vec = np.meshgrid(x_vec, y_vec)
U = -Y_vec  # Componente X
V = X_vec   # Componente Y

ax2.quiver(X_vec, Y_vec, U, V, np.sqrt(U**2 + V**2), cmap='plasma', scale=50)
ax2.set_title('Campo Vectorial', fontsize=12, fontweight='bold')
ax2.set_xlabel('X')
ax2.set_ylabel('Y')
ax2.set_xlim(-3, 3)
ax2.set_ylim(-3, 3)

plt.tight_layout()
plt.show()
````

### 4️⃣ Gráficos de Barras Horizontales, Stacked y Personalizados

````python
# Datos
categorias = ['Enero', 'Febrero', 'Marzo', 'Abril', 'Mayo']
grupo_a = [10, 24, 36, 18, 7]
grupo_b = [15, 18, 25, 30, 20]
grupo_c = [12, 14, 22, 28, 16]

fig, axes = plt.subplots(1, 3, figsize=(16, 5))

# Gráfico 1: Barras horizontales
y_pos = np.arange(len(categorias))
axes[0].barh(y_pos, grupo_a, color='#FF6B6B')
axes[0].set_yticks(y_pos)
axes[0].set_yticklabels(categorias)
axes[0].set_xlabel('Valores')
axes[0].set_title('Barras Horizontales')
axes[0].invert_yaxis()
axes[0].grid(True, alpha=0.3, axis='x')

# Gráfico 2: Barras Apiladas
x_pos = np.arange(len(categorias))
axes[1].bar(x_pos, grupo_a, label='Grupo A', color='#4ECDC4')
axes[1].bar(x_pos, grupo_b, bottom=grupo_a, label='Grupo B', color='#FFE66D')
axes[1].bar(x_pos, grupo_c, bottom=np.array(grupo_a)+np.array(grupo_b), label='Grupo C', color='#95E1D3')
axes[1].set_xticks(x_pos)
axes[1].set_xticklabels(categorias, rotation=45)
axes[1].set_ylabel('Valores')
axes[1].set_title('Barras Apiladas')
axes[1].legend()
axes[1].grid(True, alpha=0.3, axis='y')

# Gráfico 3: Barras Agrupadas
width = 0.25
axes[2].bar(x_pos - width, grupo_a, width, label='Grupo A', color='#A8D8EA')
axes[2].bar(x_pos, grupo_b, width, label='Grupo B', color='#F59E0B')
axes[2].bar(x_pos + width, grupo_c, width, label='Grupo C', color='#D97706')
axes[2].set_xticks(x_pos)
axes[2].set_xticklabels(categorias, rotation=45)
axes[2].set_ylabel('Valores')
axes[2].set_title('Barras Agrupadas')
axes[2].legend()
axes[2].grid(True, alpha=0.3, axis='y')

plt.tight_layout()
plt.show()
````

### 5️⃣ Diagramas de Violín y Distribuciones Complejas

````python
# Generar datos con distribuciones diferentes
np.random.seed(42)
datos_grupo1 = np.random.normal(100, 10, 200)
datos_grupo2 = np.random.normal(110, 15, 200)
datos_grupo3 = np.random.normal(95, 12, 200)

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Diagrama de violín
datos_violin = [datos_grupo1, datos_grupo2, datos_grupo3]
partes = axes[0].violinplot(datos_violin, positions=[1, 2, 3], showmeans=True, showmedians=True)
axes[0].set_xticks([1, 2, 3])
axes[0].set_xticklabels(['Grupo 1', 'Grupo 2', 'Grupo 3'])
axes[0].set_ylabel('Valores')
axes[0].set_title('Diagrama de Violín')
axes[0].grid(True, alpha=0.3, axis='y')

# Gráfico de caja con puntos dispersos
bp = axes[1].boxplot(datos_violin, label=['Grupo 1', 'Grupo 2', 'Grupo 3'], patch_artist=True)
colores = ['#FF6B6B', '#4ECDC4', '#95E1D3']
for patch, color in zip(bp['boxes'], colores):
    patch.set_facecolor(color)
    patch.set_alpha(0.7)

# Añadir puntos individuales
for i, datos in enumerate(datos_violin, 1):
    y = datos
    x = np.random.normal(i, 0.04, size=len(y))
    axes[1].scatter(x, y, alpha=0.3, s=20, color='gray')

axes[1].set_ylabel('Valores')
axes[1].set_title('Box Plot con Puntos Individuales')
axes[1].grid(True, alpha=0.3, axis='y')

plt.tight_layout()
plt.show()
````

### 6️⃣ Gráficos de Series Temporales con Múltiples Ejes

````python
# Generar datos de series temporales
fechas = pd.date_range('2023-01-01', periods=365, freq='D')
temperatura = 20 + 10*np.sin(np.arange(365)*2*np.pi/365) + np.random.randn(365)*2
precipitacion = np.abs(np.random.randn(365)*30)

fig, ax1 = plt.subplots(figsize=(14, 6))

# Eje 1: Temperatura
color = 'tab:red'
ax1.set_xlabel('Fecha', fontsize=12)
ax1.set_ylabel('Temperatura (°C)', color=color, fontsize=12)
line1 = ax1.plot(fechas, temperatura, color=color, linewidth=2, label='Temperatura')
ax1.tick_params(axis='y', labelcolor=color)
ax1.grid(True, alpha=0.3)

# Eje 2: Precipitación
ax2 = ax1.twinx()
color = 'tab:blue'
ax2.set_ylabel('Precipitación (mm)', color=color, fontsize=12)
line2 = ax2.bar(fechas, precipitacion, color=color, alpha=0.3, label='Precipitación', width=1)
ax2.tick_params(axis='y', labelcolor=color)

# Título y leyenda combinada
fig.suptitle('Serie Temporal: Temperatura y Precipitación (2023)', fontsize=14, fontweight='bold')
lines = line1 + [line2]
labels = [l.get_label() for l in lines]
ax1.legend(lines, labels, loc='upper left', fontsize=10)

plt.tight_layout()
plt.show()
````

### 7️⃣ Gráficos de Red y Relaciones

````python
# Crear un gráfico de correlación con círculos
np.random.seed(42)
n_variables = 6
nombres = ['Var_A', 'Var_B', 'Var_C', 'Var_D', 'Var_E', 'Var_F']

# Crear datos correlacionados
datos = np.random.randn(100, n_variables)
matriz_corr = np.corrcoef(datos.T)

fig, ax = plt.subplots(figsize=(10, 8))

# Crear heatmap de correlación
im = ax.imshow(matriz_corr, cmap='coolwarm', vmin=-1, vmax=1, aspect='auto')

# Configurar ejes
ax.set_xticks(np.arange(n_variables))
ax.set_yticks(np.arange(n_variables))
ax.set_xticklabels(nombres)
ax.set_yticklabels(nombres)
plt.setp(ax.get_xticklabels(), rotation=45, ha="right", rotation_mode="anchor")

# Añadir valores de correlación
for i in range(n_variables):
    for j in range(n_variables):
        text = ax.text(j, i, f'{matriz_corr[i, j]:.2f}',
                       ha="center", va="center", color="black" if abs(matriz_corr[i, j]) < 0.5 else "white",
                       fontsize=10, fontweight='bold')

ax.set_title('Matriz de Correlación', fontsize=14, fontweight='bold')
plt.colorbar(im, ax=ax, label='Correlación')
plt.tight_layout()
plt.show()
````

### 8️⃣ Gráficos Especiales: Radar, Polar y Más

````python
# Gráfico de radar (polar)
fig, axes = plt.subplots(1, 2, figsize=(14, 6), subplot_kw=dict(projection='polar'))

# Datos para el radar
categorias_radar = ['Velocidad', 'Defensa', 'Regate', 'Tiro', 'Pase', 'Resistencia']
n_categorias = len(categorias_radar)
angulos = np.linspace(0, 2 * np.pi, n_categorias, endpoint=False).tolist()
angulos += angulos[:1]  # Cerrar el círculo

# Datos de dos jugadores
jugador1 = [85, 78, 88, 92, 89, 85]
jugador1 += jugador1[:1]

jugador2 = [80, 90, 75, 80, 88, 90]
jugador2 += jugador2[:1]

# Gráfico 1: Un jugador
axes[0].plot(angulos, jugador1, 'o-', linewidth=2, color='#FF6B6B', label='Jugador 1')
axes[0].fill(angulos, jugador1, alpha=0.25, color='#FF6B6B')
axes[0].set_xticks(angulos[:-1])
axes[0].set_xticklabels(categorias_radar)
axes[0].set_ylim(0, 100)
axes[0].set_title('Perfil del Jugador 1', fontsize=12, fontweight='bold', pad=20)
axes[0].grid(True)

# Gráfico 2: Comparación
axes[1].plot(angulos, jugador1, 'o-', linewidth=2, color='#FF6B6B', label='Jugador 1')
axes[1].fill(angulos, jugador1, alpha=0.15, color='#FF6B6B')
axes[1].plot(angulos, jugador2, 'o-', linewidth=2, color='#4ECDC4', label='Jugador 2')
axes[1].fill(angulos, jugador2, alpha=0.15, color='#4ECDC4')
axes[1].set_xticks(angulos[:-1])
axes[1].set_xticklabels(categorias_radar)
axes[1].set_ylim(0, 100)
axes[1].set_title('Comparación de Jugadores', fontsize=12, fontweight='bold', pad=20)
axes[1].legend(loc='upper right', bbox_to_anchor=(1.3, 1.1))
axes[1].grid(True)

plt.tight_layout()
plt.show()
````
