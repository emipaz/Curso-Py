La librería Seaborn proporciona una variedad de estilos de gráficos predefinidos que se pueden utilizar para personalizar la apariencia de los gráficos. Estos estilos incluyen:

* **darkgrid**: Este estilo utiliza una cuadrícula oscura y un fondo blanco.
* **whitegrid**: Este estilo utiliza una cuadrícula blanca y un fondo oscuro.
* **dark**: Este estilo utiliza un fondo oscuro y texto blanco.
* **white**: Este estilo utiliza un fondo blanco y texto negro.
* **ticks**: Este estilo utiliza marcas de graduación en los ejes y un fondo blanco.
* **axes**: Este estilo utiliza ejes sin marcas de graduación y un fondo blanco.
* **notebook**: Este estilo está diseñado para su uso en cuadernos Jupyter y utiliza un fondo blanco y texto negro.
* **paper**: Este estilo está diseñado para su uso en documentos impresos y utiliza un fondo blanco y texto negro.
* **poster**: Este estilo está diseñado para su uso en carteles y utiliza un fondo blanco y texto negro.
* **presentation**: Este estilo está diseñado para su uso en presentaciones y utiliza un fondo blanco y texto negro.

Para establecer el estilo de un gráfico, se puede utilizar la función  `set_theme()`  de Seaborn. Por ejemplo, para establecer el estilo de un gráfico como "darkgrid", se puede utilizar el siguiente código:


```python
import seaborn as sns

sns.set_theme(style="darkgrid")


data = {"A": [1, 2, 3], "B": [4, 5, 6]}
df = pd.DataFrame(data)

sns.barplot(data=df, x="A", y="B")
plt.show()
```

La librería Seaborn proporciona una variedad de conjuntos de datos de ejemplo que se pueden utilizar para explorar y visualizar datos. Estos conjuntos de datos incluyen:

* **anscombe**: Este conjunto de datos contiene cuatro conjuntos de datos diferentes, cada uno de los cuales consta de 11 puntos de datos. Los conjuntos de datos se utilizan a menudo para ilustrar los efectos de los valores atípicos en los análisis estadísticos.
* **iris**: Este conjunto de datos contiene mediciones de tres especies diferentes de iris. Las mediciones incluyen la longitud del sépalo, la anchura del sépalo, la longitud del pétalo y la anchura del pétalo.
* **tips**: Este conjunto de datos contiene información sobre las propinas que se dan en un restaurante. La información incluye el tamaño de la cuenta, el sexo del cliente, el día de la semana y la hora del día.
* **titanic**: Este conjunto de datos contiene información sobre los pasajeros del Titanic. La información incluye el nombre del pasajero, la edad, el sexo, la clase social y si el pasajero sobrevivió o no.
* **penguins**: Este conjunto de datos contiene información sobre tres especies diferentes de pingüinos. La información incluye la especie del pingüino, la longitud del pico, la anchura del pico y la longitud de las aletas.
* **flights**: Este conjunto de datos contiene información sobre los vuelos de una aerolínea. La información incluye el número de vuelo, la fecha del vuelo, el origen del vuelo, el destino del vuelo, la distancia del vuelo y el tiempo de vuelo.
* **mpg**: Este conjunto de datos contiene información sobre el consumo de combustible de los vehículos. La información incluye el nombre del vehículo, el año del vehículo, el tipo de vehículo, el consumo de combustible en ciudad y el consumo de combustible en carretera.
* **planets**: Este conjunto de datos contiene información sobre los planetas del sistema solar. La información incluye el nombre del planeta, la masa del planeta, el radio del planeta y la distancia del planeta al Sol.

Estos conjuntos de datos se pueden utilizar para explorar y visualizar datos, así como para desarrollar y probar modelos estadísticos.

```python

# Load the example dataset for Anscombe's quartet
df = sns.load_dataset("anscombe")
df
```

```python
sns.set_theme(style="ticks")
```

```python

# Show the results of a linear regression within each dataset
sns.lmplot(
    data = df, 
    x    = "x" , 
    y    = "y", 
    col  = "dataset", 
    hue  = "dataset",
    col_wrap = 2, 
    palette = "muted", 
    ci = None,
    height = 4, 
    scatter_kws = {"s": 50, "alpha": 1}
)
```


La función  `sns.lmplot()`  de la librería Seaborn se utiliza para crear gráficos de dispersión con líneas de regresión ajustadas. La función toma una serie de argumentos, incluyendo:

*  `data` : Un DataFrame que contiene los datos que se van a graficar.
*  `x` : El nombre de la columna que se va a utilizar como variable independiente en el gráfico.
*  `y` : El nombre de la columna que se va a utilizar como variable dependiente en el gráfico.
*  `col` : El nombre de la columna que se va a utilizar para dividir los datos en grupos.
*  `hue` : El nombre de la columna que se va a utilizar para colorear los puntos de datos.
*  `col_wrap` : El número máximo de columnas en la cuadrícula de gráficos.
*  `palette` : La paleta de colores que se va a utilizar para colorear los gráficos.
*  `ci` : Si se debe mostrar o no el intervalo de confianza alrededor de las líneas de regresión.
*  `height` : La altura de cada gráfico en pulgadas.
*  `scatter_kws` : Un diccionario que contiene los atributos de los puntos de datos en los gráficos, como el tamaño ("s") y la transparencia ("alpha").

la función  `sns.lmplot()`  se utiliza para crear una cuadrícula de gráficos de dispersión con líneas de regresión ajustadas. Los datos se dividen en grupos en función de la columna "dataset", y los puntos de datos se colorean en función de la columna "dataset". La cuadrícula de gráficos tiene dos columnas y cuatro filas, y cada gráfico tiene una altura de 4 pulgadas. Los puntos de datos tienen un tamaño de 50 y una transparencia de 1.

```python
# Load the example flights dataset and convert to long-form

flights_long = sns.load_dataset("flights")
flights_long
```

```python
flights = (
    flights_long
    .pivot(index="month", 
           columns="year", 
           values="passengers")
)
flights
```

```python
# Graficamos un mapa de calor con los datos de los vuelos
f, ax = plt.subplots(figsize=(12, 10))
sns.heatmap(flights, annot=True, fmt="d", linewidths=.5, ax=ax)
```

```python
import seaborn as sns
sns.set_theme(style="ticks", palette="pastel")

# Load the example tips dataset
tips = sns.load_dataset("tips")

# Draw a nested boxplot to show bills by day and time
sns.boxplot(x="day", y="total_bill",
            hue="smoker", palette=["m", "g"],
            data=tips)
sns.despine(offset=10, trim=True)
```

```python
import seaborn as sns
sns.set_theme(style="dark")

# Load the example tips dataset
tips = sns.load_dataset("tips")

# Draw a nested violinplot and split the violins for easier comparison
sns.violinplot(data=tips, x="day", y="total_bill", hue="smoker",
               split=True, inner="quart", fill=False,
               palette={"Yes": "g", "No": ".35"})
```
