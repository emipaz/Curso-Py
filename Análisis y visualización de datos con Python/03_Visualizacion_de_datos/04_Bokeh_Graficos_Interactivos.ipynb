# 📊 Bokeh: Gráficos Interactivos para la Web

En este módulo aprenderemos a crear visualizaciones interactivas modernas usando Bokeh.

## 🤔 ¿Qué es Bokeh?

**Bokeh** es una biblioteca de Python especializada en crear **visualizaciones interactivas** que se visualizan directamente en un navegador web.

### Características principales:
- 📊 Gráficos interactivos con zoom, pan y hover
- 🌐 Exportación como archivos HTML independientes
- 🎨 Personalización completa (colores, estilos, temas)
- 📱 Dashboards responsivos
- ⚡ Excelente rendimiento incluso con muchos datos
- 🔗 Gráficos vinculados y comunicación entre visualizaciones

## 🎯 Conceptos Fundamentales

Bokeh funciona con tres componentes principales:

```
Figure (la ventana/lienzo)
  └── Glifos (Glyphs) - círculos, líneas, barras
      └── Fuente de datos (ColumnDataSource)
```

### Los tres elementos:

1. **Figure (Figura)**: El lienzo donde dibujamos
2. **Glifos (Glyphs)**: Los elementos visuales (círculos, líneas, etc.)
3. **ColumnDataSource**: Contenedor estructurado de datos

## 📈 Tu Primer Gráfico con Bokeh

````python
# Crear datos simples
x = np.linspace(0, 10, 100)
y = np.sin(x)

# Crear figura
p = figure(
    title='Mi Primer Gráfico - Función Seno',
    x_axis_label='X',
    y_axis_label='Y = sin(X)',
    width=800,
    height=400,
    toolbar_location='right',
    tools='pan,wheel_zoom,box_zoom,reset,save'
)

# Agregar línea
p.line(x, y, line_width=2, color='blue', legend_label='sin(x)')

# Personalizar
p.legend.click_policy = 'hide'  # Clic en leyenda para ocultar/mostrar

show(p)
````

## 🎨 Tipos de Gráficos Básicos

````python
# Preparar datos de ejemplo
x = np.linspace(0, 10, 50)
y = np.sin(x)
categorias = ['A', 'B', 'C', 'D', 'E']
valores = [23, 45, 56, 78, 32]

# 1. Gráfico de Línea
p1 = figure(title='Gráfico de Línea', width=350, height=350, 
             x_axis_label='X', y_axis_label='sin(X)')
p1.line(x, y, line_width=2, color='blue')
p1.circle(x, y, size=6, color='blue', alpha=0.5)

# 2. Gráfico de Barras
p2 = figure(x_range=categorias, title='Gráfico de Barras', 
             width=350, height=350, y_axis_label='Valores')
p2.vbar(x=categorias, top=valores, width=0.7, color='steelblue')

# 3. Gráfico de Dispersión
x_scatter = np.random.randn(100)
y_scatter = np.random.randn(100)
p3 = figure(title='Gráfico de Dispersión', width=350, height=350,
             x_axis_label='X', y_axis_label='Y')
p3.circle(x_scatter, y_scatter, size=8, color='navy', alpha=0.5)

# 4. Histograma
datos = np.random.normal(100, 15, 1000)
hist, edges = np.histogram(datos, density=True, bins=30)
p4 = figure(title='Histograma', width=350, height=350,
             x_axis_label='Valores', y_axis_label='Frecuencia')
p4.quad(top=hist, bottom=0, left=edges[:-1], right=edges[1:], 
         fill_color='green', line_color='white', alpha=0.5)

# 5. Gráfico de Área
x_area = np.linspace(0, 10, 100)
y1 = np.sin(x_area)
y2 = np.cos(x_area)
p5 = figure(title='Gráfico de Área', width=350, height=350,
             x_axis_label='X', y_axis_label='Y')
p5.varea(x=x_area, y1=y1, y2=0, alpha=0.3, color='blue', legend_label='sin(x)')
p5.line(x_area, y1, line_width=2, color='blue')
p5.line(x_area, y2, line_width=2, color='red', legend_label='cos(x)')
p5.legend.click_policy = 'hide'

# 6. Barras Horizontales
p6 = figure(y_range=categorias, title='Barras Horizontales',
             width=350, height=350, x_axis_label='Valores')
p6.hbar(y=categorias, right=valores, height=0.7, color='coral')

# Organizar en grid
grid = gridplot([[p1, p2, p3], [p4, p5, p6]])
show(grid)
````

## 🎯 Personalización Básica

````python
# Crear un gráfico con personalización completa
meses = ['Ene', 'Feb', 'Mar', 'Abr', 'May', 'Jun', 'Jul', 'Ago', 'Sep', 'Oct', 'Nov', 'Dic']
ventas_a = [10, 12, 15, 18, 22, 25, 23, 26, 28, 32, 35, 38]
ventas_b = [8, 10, 12, 14, 16, 18, 19, 20, 22, 24, 26, 28]

# Crear figura personalizada
p = figure(
    x_range=meses,
    title='Comparación de Ventas Mensuales',
    x_axis_label='Mes',
    y_axis_label='Ventas (en miles)',
    width=900,
    height=450,
    toolbar_location='right',
    tools='pan,wheel_zoom,box_zoom,reset,save',
    background_fill_color='#f5f5f5'
)

# Posiciones para barras lado a lado
x_offset_1 = [-0.2] * len(meses)
x_offset_2 = [0.2] * len(meses)

# Agregar barras
p.vbar(x=[i + x_offset_1[i] for i in range(len(meses))], top=ventas_a, 
        width=0.35, color='#2E86AB', legend_label='Producto A', alpha=0.8)
p.vbar(x=[i + x_offset_2[i] for i in range(len(meses))], top=ventas_b, 
        width=0.35, color='#A23B72', legend_label='Producto B', alpha=0.8)

# Personalizar
p.legend.location = 'top_left'
p.legend.click_policy = 'hide'
p.xaxis.axis_label_text_font_size = '12pt'
p.yaxis.axis_label_text_font_size = '12pt'
p.title.text_font_size = '14pt'

show(p)
````

## 🎛️ Herramientas Interactivas: Hover Tool

````python
# Crear datos
datos = pd.DataFrame({
    'Mes': ['Ene', 'Feb', 'Mar', 'Abr', 'May', 'Jun'],
    'Ventas': [100, 125, 110, 145, 160, 155],
    'Región': ['Norte', 'Sur', 'Norte', 'Este', 'Oeste', 'Centro']
})

# Convertir a ColumnDataSource
source = ColumnDataSource(datos)

# Crear figura
p = figure(
    x_range=datos['Mes'].tolist(),
    title='Ventas Mensuales con Hover Interactivo',
    width=800,
    height=400,
    toolbar_location='right',
    tools='pan,wheel_zoom,box_zoom,reset,save,hover'
)

# Agregar barras
p.vbar(x='Mes', top='Ventas', width=0.7, source=source, 
        color='#3182bd', alpha=0.8)

# Configurar hover tool
hover = p.select_one(HoverTool)
hover.tooltips = [
    ('Mes', '@Mes'),
    ('Ventas', '@Ventas{0,0}'),
    ('Región', '@Región')
]

show(p)
````

## 🎨 Paletas de Colores y Personalización

````python
# Crear datos de múltiples series
datos = pd.DataFrame({
    'Categoría': ['A', 'B', 'C', 'D', 'E', 'A', 'B', 'C', 'D', 'E'],
    'Valores': [20, 30, 25, 35, 40, 25, 35, 30, 40, 45],
    'Grupo': ['Grupo 1', 'Grupo 1', 'Grupo 1', 'Grupo 1', 'Grupo 1',
              'Grupo 2', 'Grupo 2', 'Grupo 2', 'Grupo 2', 'Grupo 2']
})

# Obtener paleta de colores
colors = Category10[5]

# Crear figura (sin x_range categórico)
p = figure(
    title='Gráfico con Paletas de Colores',
    width=800,
    height=400,
    toolbar_location='right'
)

# Barras con colores de paleta
grupo1 = datos[datos['Grupo'] == 'Grupo 1']
grupo2 = datos[datos['Grupo'] == 'Grupo 2']

p.vbar(x=[i - 0.2 for i in range(len(grupo1))], top=grupo1['Valores'].values,
        width=0.35, color=colors[0], legend_label='Grupo 1', alpha=0.8)
p.vbar(x=[i + 0.2 for i in range(len(grupo2))], top=grupo2['Valores'].values,
        width=0.35, color=colors[1], legend_label='Grupo 2', alpha=0.8)

# Personalizar ejes (esto ahora sí tiene sentido, porque el eje es numérico)
p.xaxis.ticker = list(range(len(grupo1)))
p.xaxis.major_label_overrides = dict(zip(range(len(grupo1)), grupo1['Categoría'].values))
p.legend.location = 'top_left'

show(p)
````

## 📊 Scatter Plot Avanzado

````python
# Crear datos de iris
iris = pd.read_csv('https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv')

# Mapear especies a colores
color_map = {'setosa': '#e74c3c', 'versicolor': '#3498db', 'virginica': '#2ecc71'}
iris['color'] = iris['species'].map(color_map)

# Convertir a ColumnDataSource
source = ColumnDataSource(iris)

# Crear figura
p = figure(
    title='Scatter Plot: Sepal Length vs Sepal Width',
    x_axis_label='Sepal Length (cm)',
    y_axis_label='Sepal Width (cm)',
    width=800,
    height=500,
    toolbar_location='right',
    tools='pan,wheel_zoom,box_zoom,reset,save,hover'
)

# Agregar círculos con tamaño basado en petal_length
p.circle('sepal_length', 'sepal_width', size=8, source=source,
          color='color', alpha=0.6, legend_field='species')

# Configurar hover
hover = p.select_one(HoverTool)
hover.tooltips = [
    ('Species', '@species'),
    ('Sepal Length', '@sepal_length{0.0}'),
    ('Sepal Width', '@sepal_width{0.0}'),
    ('Petal Length', '@petal_length{0.0}')
]

p.legend.location = 'top_right'
p.legend.click_policy = 'hide'

show(p)
````

## 🔗 Gráficos Múltiples en Layouts

````python
# Preparar datos
x = np.linspace(0, 10, 100)

# Gráfico 1: Seno
p1 = figure(title='sin(x)', width=400, height=350, 
             tools='pan,wheel_zoom,reset')
p1.line(x, np.sin(x), line_width=2, color='blue')
p1.circle(x[::10], np.sin(x)[::10], size=6, color='blue', alpha=0.5)

# Gráfico 2: Coseno
p2 = figure(title='cos(x)', width=400, height=350,
             tools='pan,wheel_zoom,reset')
p2.line(x, np.cos(x), line_width=2, color='red')
p2.circle(x[::10], np.cos(x)[::10], size=6, color='red', alpha=0.5)

# Gráfico 3: Tangente
p3 = figure(title='tan(x)', width=400, height=350,
             y_range=(-5, 5), tools='pan,wheel_zoom,reset')
p3.line(x, np.tan(x), line_width=2, color='green')

# Gráfico 4: Exponencial
p4 = figure(title='exp(x/2)', width=400, height=350,
             tools='pan,wheel_zoom,reset')
p4.line(x, np.exp(x/2), line_width=2, color='purple')

# Organizar en layout
layout = column(
    row(p1, p2),
    row(p3, p4)
)

show(layout)
````

## 🎯 Anotaciones en Gráficos

````python
from bokeh.models import Label, LabelSet

# Crear datos
x = np.array([1, 2, 3, 4, 5])
y = np.array([10, 15, 13, 17, 20])

source = ColumnDataSource(data=dict(
    x=x, y=y,
    etiqueta=['Punto 1', 'Punto 2', 'Punto 3', 'Punto 4', 'Punto 5']
))

# Crear figura
p = figure(
    title='Gráfico con Anotaciones',
    width=800,
    height=400,
    toolbar_location='right'
)

# Agregar puntos
p.circle('x', 'y', size=10, source=source, color='navy', alpha=0.6)
p.line(x, y, line_width=2, color='navy')

# Agregar etiquetas
labels = LabelSet(x='x', y='y', text='etiqueta', source=source,
                   x_offset=5, y_offset=5, text_font_size='10pt')
p.add_layout(labels)

show(p)
````
