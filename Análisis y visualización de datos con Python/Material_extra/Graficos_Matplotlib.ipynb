# Ejemplos de Gráficos Clásicos con Matplotlib

Este notebook muestra cómo crear los gráficos más comunes usando Matplotlib en Python.

## 1. Importar Librerías
Importamos Matplotlib y numpy para los ejemplos.

```python
import matplotlib.pyplot as plt
import numpy as np
plt.style.use('seaborn-v0_8')
print("Matplotlib version:", plt.matplotlib.__version__)
```

## 2. Gráfico de Línea
El gráfico más clásico para mostrar tendencias.

```python
x = np.arange(0, 10, 1)
y = x ** 2
plt.plot(x, y, marker='o', color='blue')
plt.title('Gráfico de Línea')
plt.xlabel('Eje X')
plt.ylabel('Eje Y')
plt.grid(True)
plt.show()
```

## 3. Gráfico de Barras
Ideal para comparar cantidades entre categorías.

```python
categorias = ['A', 'B', 'C', 'D']
valores = [23, 45, 12, 36]
plt.bar(categorias, valores, color='orange')
plt.title('Gráfico de Barras')
plt.xlabel('Categoría')
plt.ylabel('Valor')
plt.show()
```

## 4. Gráfico de Dispersión (Scatter)
Útil para analizar la relación entre dos variables.

```python
np.random.seed(0)
x = np.random.rand(50)
y = 2 * x + np.random.normal(0, 0.1, 50)
plt.scatter(x, y, color='green')
plt.title('Gráfico de Dispersión')
plt.xlabel('Variable X')
plt.ylabel('Variable Y')
plt.show()
```

## 5. Histograma
Para visualizar la distribución de una variable numérica.

```python
datos = np.random.normal(0, 1, 1000)
plt.hist(datos, bins=30, color='purple', edgecolor='black', alpha=0.7)
plt.title('Histograma')
plt.xlabel('Valor')
plt.ylabel('Frecuencia')
plt.show()
```

## 6. Gráfico de Pastel (Pie Chart)
Para mostrar proporciones de un total.

```python
labels = ['Manzanas', 'Bananas', 'Cerezas', 'Dátiles']
sizes = [30, 25, 25, 20]
plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=90, colors=['red', 'yellow', 'pink', 'brown'])
plt.title('Gráfico de Pastel')
plt.axis('equal')
plt.show()
```

## 7. Boxplot
Para visualizar la dispersión y posibles outliers de una variable.

```python
datos = np.random.normal(50, 15, 200)
plt.boxplot(datos, patch_artist=True, boxprops=dict(facecolor='lightblue'))
plt.title('Boxplot')
plt.ylabel('Valor')
plt.show()
```

## 8. Anotaciones, Líneas y Elementos en Matplotlib

Puedes agregar anotaciones, líneas verticales/horizontales y otros elementos para resaltar información en tus gráficos.

```python
# Ejemplo: agregar anotaciones y líneas a un gráfico de línea
x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y, label='Seno(x)')

# Línea vertical en x=5
ymax = y.max()
plt.axvline(x=5, color='red', linestyle='--', label='x=5')

# Línea horizontal en y=0
plt.axhline(y=0, color='gray', linestyle=':')

# Anotación en el máximo
plt.annotate('Máximo', xy=(x[np.argmax(y)], ymax), xytext=(x[np.argmax(y)]+1, ymax),
             arrowprops=dict(facecolor='black', shrink=0.05), fontsize=12, color='blue')

# Agregar una caja de texto
plt.text(7, -0.8, 'Ejemplo de texto', bbox=dict(facecolor='yellow', alpha=0.3))

plt.title('Gráfico con Anotaciones y Líneas')
plt.xlabel('x')
plt.ylabel('sin(x)')
plt.legend()
plt.grid(True)
plt.show()
```

## 9. Otros Elementos Útiles en Matplotlib

A continuación, algunos elementos adicionales que puedes agregar a tus gráficos para enriquecer la visualización:

```python
# Flechas personalizadas
x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)
plt.plot(x, y)

# Flecha entre dos puntos
plt.annotate('', xy=(np.pi, 0), xytext=(np.pi/2, 1), arrowprops=dict(arrowstyle='->', color='red', lw=2))
plt.text(np.pi/2, 1.1, 'Flecha personalizada', color='red')

plt.title('Flechas personalizadas en el gráfico')
plt.show()

# Sombreado de áreas
x = np.linspace(0, 10, 200)
y = np.cos(x)
plt.plot(x, y)

# Sombrear entre x=2 y x=6
plt.axvspan(2, 6, color='yellow', alpha=0.3, label='Zona destacada')
plt.legend()
plt.title('Sombreado de áreas con axvspan')
plt.show()

# Sombrear bajo la curva
y2 = np.exp(-0.1 * x) * np.sin(x)
plt.plot(x, y2)
plt.fill_between(x, y2, color='lightgreen', alpha=0.5)
plt.title('Sombreado bajo la curva')
plt.show()

# Agregar múltiples líneas de referencia
plt.plot(x, y2)
for val in [2, 4, 6, 8]:
    plt.axvline(x=val, color='gray', linestyle='--', alpha=0.5)
plt.title('Múltiples líneas de referencia')
plt.show()
```
