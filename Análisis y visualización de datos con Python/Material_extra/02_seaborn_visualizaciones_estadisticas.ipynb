# 📊 Seaborn: Visualizaciones Estadísticas Avanzadas

Bienvenidos a Seaborn. Esta librería está construida sobre Matplotlib y hace que crear visualizaciones estadísticas hermosas y complejas sea muy simple.

## 🤔 ¿Qué es Seaborn?

**Seaborn** es una librería de Python basada en Matplotlib que proporciona una interfaz de alto nivel para crear visualizaciones estadísticas atractivas.

### Diferencias con Matplotlib:
| Aspecto | Matplotlib | Seaborn |
|--------|-----------|----------|
| **Nivel** | Bajo (control total) | Alto (simplificado) |
| **Estética** | Básica | Elegante por defecto |
| **Temas** | Pocos estilos | Múltiples temas |
| **Integración** | Necesita pandas manualmente | Integrada con pandas |
| **Estadística** | Manual | Automática |
| **Uso** | Scripts complejos | Análisis exploratorio |

### Cuándo usar cada una:
- **Matplotlib**: Cuando necesitas control total o publicaciones científicas
- **Seaborn**: Para análisis exploratorio de datos y visualizaciones estadísticas
- **Ambas**: Seaborn usa Matplotlib internamente, puedes combinarlas

## 📜 Breve Historia

**Creador:** Michael Waskom  
**Año:** 2012  
**Motivación:** Simplificar la creación de visualizaciones estadísticas en Matplotlib, que requería mucho código boilerplate.

Seaborn se convirtió rápidamente en estándar para análisis exploratorio de datos en Python.

## 🛠️ Instalación y Setup

```python
# !pip install seaborn pandas numpy matplotlib

import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Configurar estilo
%matplotlib inline

# Verificar versión
print(f"Versión de Seaborn: {sns.__version__}")
print(f"Versión de Pandas: {pd.__version__}")
```

## 🎨 Temas y Estilos de Seaborn

```python
# Ver temas disponibles
temas = sns.axes_style().keys()
print("Estilos disponibles en Seaborn:")
print(sns.available_themes())

# Ver contextos disponibles
print("\nContextos disponibles:")
print(['paper', 'notebook', 'talk', 'poster'])
```

```python
# Demostración de diferentes temas
fig, axes = plt.subplots(2, 3, figsize=(15, 10))
fig.suptitle('Diferentes Temas en Seaborn', fontsize=16, fontweight='bold')

temas = ['darkgrid', 'whitegrid', 'dark', 'white', 'ticks']
datos = np.random.randn(100, 2)

for idx, (ax, tema) in enumerate(zip(axes.flat[:5], temas)):
    sns.set_style(tema)
    sns.scatterplot(x=datos[:, 0], y=datos[:, 1], ax=ax, s=100)
    ax.set_title(f'Tema: {tema}', fontweight='bold')

# Mostrar paleta de colores
sns.set_style('whitegrid')
color_palettes = sns.color_palette('husl', 8)
axes.flat[5].bar(range(8), [1]*8, color=color_palettes)
axes.flat[5].set_title('Paleta: Husl', fontweight='bold')
axes.flat[5].set_xticks([])

plt.tight_layout()
plt.show()
```

## 📈 Gráficos Principales de Seaborn

### 1. Relaciones entre Variables: scatterplot y lineplot

```python
# Cargar dataset de ejemplo
tips = sns.load_dataset('tips')
print(tips.head())
print(f"\nShape: {tips.shape}")
```

```python
# Scatter plot con colores y tamaños
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Scatter plot básico
sns.scatterplot(data=tips, x='total_bill', y='tip', 
                hue='sex', size='size', 
                palette='Set2', ax=axes[0])
axes[0].set_title('Relación entre Total Bill y Tip', fontweight='bold')
axes[0].set_xlabel('Total Bill ($)')
axes[0].set_ylabel('Tip ($)')

# Line plot por día
flight_data = sns.load_dataset('flights')
sns.lineplot(data=flight_data, x='month', y='passengers', 
             hue='year', palette='tab10', ax=axes[1])
axes[1].set_title('Pasajeros por Mes (1949-1960)', fontweight='bold')
axes[1].set_xlabel('Mes')
axes[1].set_ylabel('Número de Pasajeros')

plt.tight_layout()
plt.show()
```

### 2. Distribuciones: histplot, kdeplot, distplot

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('Gráficos de Distribución en Seaborn', fontsize=14, fontweight='bold')

# Histograma
sns.histplot(data=tips, x='total_bill', kde=True, ax=axes[0, 0], hue='sex')
axes[0, 0].set_title('Histograma con KDE')

# KDE plot
sns.kdeplot(data=tips, x='total_bill', y='tip', ax=axes[0, 1], fill=True, cmap='Blues')
axes[0, 1].set_title('KDE Bivariado')

# Violin plot
sns.violinplot(data=tips, x='day', y='total_bill', hue='sex', ax=axes[1, 0])
axes[1, 0].set_title('Violin Plot')

# Rug plot
sns.histplot(data=tips, x='total_bill', ax=axes[1, 1])
sns.rugplot(data=tips, x='total_bill', ax=axes[1, 1], height=0.05, color='red')
axes[1, 1].set_title('Histograma con Rug Plot')

plt.tight_layout()
plt.show()
```

### 3. Comparaciones Categóricas: barplot, boxplot, stripplot

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('Gráficos Categóricos en Seaborn', fontsize=14, fontweight='bold')

# Bar plot
sns.barplot(data=tips, x='day', y='total_bill', hue='sex', ax=axes[0, 0])
axes[0, 0].set_title('Bar Plot')

# Box plot
sns.boxplot(data=tips, x='day', y='total_bill', hue='sex', ax=axes[0, 1])
axes[0, 1].set_title('Box Plot')

# Violin plot
sns.violinplot(data=tips, x='day', y='total_bill', ax=axes[1, 0])
axes[1, 0].set_title('Violin Plot')

# Strip plot con jitter
sns.stripplot(data=tips, x='day', y='total_bill', hue='sex', 
              jitter=True, size=6, alpha=0.6, ax=axes[1, 1])
axes[1, 1].set_title('Strip Plot con Jitter')

plt.tight_layout()
plt.show()
```

### 4. Matrices de Datos: heatmap, clustermap

```python
# Crear matriz de correlación
iris = sns.load_dataset('iris')
corr_matrix = iris.corr(numeric_only=True)

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Heatmap simple
sns.heatmap(corr_matrix, annot=True, fmt='.2f', cmap='coolwarm', 
            center=0, ax=axes[0], cbar_kws={'label': 'Correlación'})
axes[0].set_title('Matriz de Correlación (Iris)')

# Heatmap con máscara (triángulo)
mask = np.triu(np.ones_like(corr_matrix, dtype=bool))
sns.heatmap(corr_matrix, mask=mask, annot=True, fmt='.2f', 
            cmap='RdBu_r', ax=axes[1], cbar_kws={'label': 'Correlación'})
axes[1].set_title('Matriz de Correlación (Triángulo Inferior)')

plt.tight_layout()
plt.show()
```

### 5. Gráficos Multivariados: pairplot y FacetGrid

```python
# Pair plot - matriz de dispersión
pairplot = sns.pairplot(iris, hue='species', palette='Set2', diag_kind='kde')
pairplot.fig.suptitle('Pair Plot del Dataset Iris', y=1.001, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
# FacetGrid - subgráficos por categoría
g = sns.FacetGrid(tips, col='time', row='sex', height=4, aspect=1.2)
g.map(sns.scatterplot, 'total_bill', 'tip', s=100, alpha=0.6)
g.fig.suptitle('Tips por Tiempo y Sexo', y=1.001, fontweight='bold')
plt.tight_layout()
plt.show()
```

### 6. Gráficos de Regresión: regplot y lmplot

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Regplot simple
sns.regplot(data=tips, x='total_bill', y='tip', ax=axes[0], 
            scatter_kws={'alpha': 0.6}, line_kws={'color': 'red', 'linewidth': 2})
axes[0].set_title('Regplot: Relación Lineal')

# Regplot con orden polinómico
sns.regplot(data=tips, x='total_bill', y='tip', ax=axes[1], 
            order=2, scatter_kws={'alpha': 0.6}, 
            line_kws={'color': 'green', 'linewidth': 2})
axes[1].set_title('Regplot: Regresión Polinómica (orden 2)')

plt.tight_layout()
plt.show()
```

```python
# lmplot - regresión por categorías
lm = sns.lmplot(data=tips, x='total_bill', y='tip', hue='sex', 
                 col='time', height=4, aspect=1.2)
lm.fig.suptitle('Regresión de Tips por Tiempo y Sexo', y=1.001, fontweight='bold')
plt.tight_layout()
plt.show()
```

## 🎛️ Widgets Interactivos con Seaborn

```python
from ipywidgets import interact, Dropdown, IntSlider, FloatSlider

@interact
def explorador_seaborn(
    grafico_tipo=Dropdown(options=['scatter', 'kde', 'hist', 'violin', 'box'], value='scatter'),
    variable_x=Dropdown(options=['total_bill', 'tip', 'size', 'day'], value='total_bill'),
    variable_y=Dropdown(options=['tip', 'total_bill', 'size'], value='tip'),
    hue=Dropdown(options=[None, 'sex', 'day', 'time', 'smoker'], value='sex')
):
    
    plt.figure(figsize=(12, 6))
    
    if grafico_tipo == 'scatter':
        sns.scatterplot(data=tips, x=variable_x, y=variable_y, hue=hue, s=100, alpha=0.6)
    elif grafico_tipo == 'kde':
        if hue is None:
            sns.kdeplot(data=tips, x=variable_x, y=variable_y, fill=True, cmap='Blues')
        else:
            sns.kdeplot(data=tips, x=variable_x, y=variable_y, hue=hue, fill=True)
    elif grafico_tipo == 'hist':
        sns.histplot(data=tips, x=variable_x, hue=hue, kde=True)
    elif grafico_tipo == 'violin':
        sns.violinplot(data=tips, x=variable_x, y=variable_y, hue=hue)
    elif grafico_tipo == 'box':
        sns.boxplot(data=tips, x=variable_x, y=variable_y, hue=hue)
    
    plt.title(f'{grafico_tipo.capitalize()}: {variable_x} vs {variable_y}', fontweight='bold', fontsize=12)
    plt.tight_layout()
    plt.show()
```

## 🌟 Paletas de Colores en Seaborn

```python
# Mostrar diferentes paletas
paletas = ['deep', 'muted', 'pastel', 'bright', 'dark', 'colorblind', 'husl', 'Set2']

fig, axes = plt.subplots(len(paletas), 1, figsize=(12, 10))
fig.suptitle('Paletas de Color en Seaborn', fontsize=14, fontweight='bold')

for ax, paleta in zip(axes, paletas):
    colors = sns.color_palette(paleta, 8)
    ax.bar(range(8), [1]*8, color=colors)
    ax.set_title(f'Paleta: {paleta}', fontweight='bold')
    ax.set_xticks([])
    ax.set_ylim(0, 1.2)

plt.tight_layout()
plt.show()
```

## 📊 Ejemplo Completo: Análisis Exploratorio de Datos

```python
# Análisis completo del dataset tips
penguins = sns.load_dataset('penguins')
print(penguins.info())
print("\nPrimeras filas:")
print(penguins.head())
```

```python
# Crear un análisis exploratorio completo
fig = plt.figure(figsize=(16, 12))
gs = fig.add_gridspec(3, 3, hspace=0.3, wspace=0.3)

# 1. Distribución de masa corporal
ax1 = fig.add_subplot(gs[0, 0])
sns.histplot(data=penguins, x='body_mass_g', kde=True, ax=ax1, color='steelblue')
ax1.set_title('Distribución de Masa Corporal')

# 2. Masa corporal por especie
ax2 = fig.add_subplot(gs[0, 1])
sns.violinplot(data=penguins, x='species', y='body_mass_g', ax=ax2, palette='Set2')
ax2.set_title('Masa Corporal por Especie')

# 3. Relación longitud-masa
ax3 = fig.add_subplot(gs[0, 2])
sns.scatterplot(data=penguins, x='bill_length_mm', y='body_mass_g', 
                hue='species', ax=ax3, s=100, alpha=0.6)
ax3.set_title('Longitud Bill vs Masa Corporal')

# 4. Box plot por sexo
ax4 = fig.add_subplot(gs[1, 0])
sns.boxplot(data=penguins, x='sex', y='body_mass_g', hue='species', ax=ax4)
ax4.set_title('Masa Corporal por Sexo y Especie')

# 5. Densidad bivariada
ax5 = fig.add_subplot(gs[1, 1])
sns.kdeplot(data=penguins, x='bill_length_mm', y='flipper_length_mm', 
            fill=True, cmap='Blues', ax=ax5)
ax5.set_title('Densidad: Bill vs Flipper')

# 6. Matriz de correlación
ax6 = fig.add_subplot(gs[1, 2])
corr = penguins.corr(numeric_only=True)
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', center=0, ax=ax6, cbar=False)
ax6.set_title('Matriz de Correlación')

# 7. Conteo por especie y sexo
ax7 = fig.add_subplot(gs[2, 0])
sns.countplot(data=penguins, x='species', hue='sex', ax=ax7, palette='husl')
ax7.set_title('Conteo de Pingüinos por Especie y Sexo')

# 8. Regresión
ax8 = fig.add_subplot(gs[2, 1])
sns.regplot(data=penguins, x='flipper_length_mm', y='body_mass_g', ax=ax8,
           scatter_kws={'alpha': 0.5}, line_kws={'color': 'red', 'linewidth': 2})
ax8.set_title('Regresión: Flipper vs Masa')

# 9. Pairplot reducido (solo 2x2)
ax9 = fig.add_subplot(gs[2, 2])
sns.scatterplot(data=penguins, x='bill_depth_mm', y='flipper_length_mm',
                hue='species', size='body_mass_g', ax=ax9, alpha=0.6)
ax9.set_title('Bill Depth vs Flipper (tamaño = masa)')

fig.suptitle('Análisis Exploratorio: Dataset Pingüinos', fontsize=16, fontweight='bold', y=0.995)
plt.show()
```
