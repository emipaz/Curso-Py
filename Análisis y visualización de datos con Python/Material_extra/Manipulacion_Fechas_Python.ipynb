## 3. Funciones Básicas de Datetime {#3}

Las funciones principales para manipular objetos datetime son `strptime()` (string to datetime) y `strftime()` (datetime to string):

```python
# 1. strptime() - Convertir string a datetime
fecha_string = "25/11/2022"
dt_object = datetime.strptime(fecha_string, "%d/%m/%Y")
print(f"String: '{fecha_string}' → Datetime: {dt_object}")
print(f"Tipo: {type(dt_object)}")
print()
```

```python

# 2. strftime() - Convertir datetime a string
fecha_formateada = dt_object.strftime("%d/%m/%Y")
print(f"Datetime: {dt_object} → String: '{fecha_formateada}'")
print(f"Tipo: {type(fecha_formateada)}")
print()
```

```python

# 3. timestamp() - Convertir a timestamp Unix
timestamp_unix = dt_object.timestamp()
print(f"Datetime: {dt_object} → Timestamp: {timestamp_unix}")
print(f"Tipo: {type(timestamp_unix)}")
print()
```

```python

# 4. fromtimestamp() - Convertir timestamp a datetime
dt_from_timestamp = datetime.fromtimestamp(timestamp_unix)
print(f"Timestamp: {timestamp_unix} → Datetime: {dt_from_timestamp}")
print()
```

## 4. Conversión entre Formatos de 12 y 24 Horas {#4}

Ejemplos de conversión entre formatos de hora:

```python
# Conversión de 24 horas a 12 horas (AM/PM)
hora_24 = "20:00"
hora_12 = datetime.strptime(hora_24, "%H:%M").strftime("%I:%M %p")
print(f"24 horas: {hora_24} → 12 horas: {hora_12}")
```

```python

# Conversión de 12 horas (AM/PM) a 24 horas
hora_pm = "08:00 PM"
hora_24_convertida = datetime.strptime(hora_pm, "%I:%M %p").strftime("%H:%M")
print(f"12 horas: {hora_pm} → 24 horas: {hora_24_convertida}")
```

```python

# Más ejemplos
ejemplos_conversion = [
    ("14:30", "%H:%M", "%I:%M %p"),
    ("02:15 AM", "%I:%M %p", "%H:%M"),
    ("11:45 PM", "%I:%M %p", "%H:%M"),
    ("09:00", "%H:%M", "%I:%M %p")
]

print("\n🔄 Tabla de conversiones:")
print(f"{'Entrada':<15} {'Formato Entrada':<15} {'Salida':<15} {'Formato Salida'}")
print("-" * 65)

for entrada, fmt_entrada, fmt_salida in ejemplos_conversion:
    salida = datetime.strptime(entrada, fmt_entrada).strftime(fmt_salida)
    print(f"{entrada:<15} {fmt_entrada:<15} {salida:<15} {fmt_salida}")
```

## 5. Manipulación de Zonas Horarias {#5}

Trabajar con diferentes zonas horarias usando `pytz`:

```python
# Ejemplo del documento: Nueva York a Tokyo
ny_time_str = "25-11-2022 09:34:00-0700"
ny_time = datetime.strptime(ny_time_str, "%d-%m-%Y %H:%M:%S%z")
print(f"Hora en Nueva York: {ny_time}")
```

```python

# Convertir a Tokyo
tokyo_tz = pytz_timezone('Asia/Tokyo')
tokyo_time = ny_time.astimezone(tokyo_tz)
print(f"Hora en Tokyo: {tokyo_time}")
print()
```

```python

# Trabajar con zonas horarias comunes
zonas_horarias = ['UTC', 'US/Eastern', 'US/Pacific', 'Europe/London', 
                  'Europe/Madrid', 'Asia/Tokyo', 'Australia/Sydney', 'America/Buenos_Aires']

# Crear fecha con timezone UTC
utc_time = datetime.now(pytz.UTC)
print(f"Hora UTC actual: {utc_time}")
print("\n🌍 Conversión a diferentes zonas horarias:")
print("-" * 50)

for zona in zonas_horarias:
    tz = pytz_timezone(zona)
    hora_local = utc_time.astimezone(tz)
    print(f"{zona:<20}: {hora_local.strftime('%Y-%m-%d %H:%M:%S %Z')}")

# Ejemplo práctico: Programar una reunión internacional
print("\n📅 Ejemplo: Reunión a las 15:00 hora de Madrid")
madrid_tz = pytz_timezone('Europe/Madrid')
reunion_madrid = madrid_tz.localize(datetime(2024, 3, 15, 15, 0, 0))

print(f"Madrid: {reunion_madrid.strftime('%H:%M %Z')}")
print(f"Nueva York: {reunion_madrid.astimezone(pytz_timezone('US/Eastern')).strftime('%H:%M %Z')}")
print(f"Tokyo: {reunion_madrid.astimezone(pytz_timezone('Asia/Tokyo')).strftime('%H:%M %Z')}")
print(f"Los Angeles: {reunion_madrid.astimezone(pytz_timezone('US/Pacific')).strftime('%H:%M %Z')}")
print(f"Buenos Aires: {reunion_madrid.astimezone(pytz_timezone('America/Buenos_Aires')).strftime('%H:%M %Z')}")
```

```python

# Operaciones útiles con timestamps
print("📊 Operaciones con timestamps:")

# Calcular diferencia en días
fecha1 = datetime(2024, 1, 1)
fecha2 = datetime(2024, 12, 31)
timestamp1 = fecha1.timestamp()
timestamp2 = fecha2.timestamp()

diferencia_segundos = timestamp2 - timestamp1
diferencia_dias = diferencia_segundos / (24 * 60 * 60)

print(f"Fecha 1: {fecha1.strftime('%d/%m/%Y')}")
print(f"Fecha 2: {fecha2.strftime('%d/%m/%Y')}")
print(f"Diferencia en segundos: {diferencia_segundos:,.0f}")
print(f"Diferencia en días: {diferencia_dias:,.0f}")
```

```python

# Timestamps para diferentes momentos del día
print("\n⏰ Timestamps para diferentes horas:")
base_fecha = datetime(2024, 6, 15)
horas = [0, 6, 12, 18, 23]

for hora in horas:
    fecha_hora = base_fecha.replace(hour=hora)
    timestamp = fecha_hora.timestamp()
    print(f"{fecha_hora.strftime('%H:%M')}: {timestamp:,.0f}")
```

## 7. Datetime en NumPy {#7}

NumPy proporciona tipos `datetime64` y `timedelta64` para operaciones eficientes con arrays de fechas:

```python
# Crear datetime64 en NumPy
fecha_numpy = np.datetime64('2024-03-15')
print(f"Fecha NumPy: {fecha_numpy}")
print(f"Tipo: {type(fecha_numpy)}")
print()
```

```python

# Crear array de fechas
fechas_array = np.array(['2024-01-01', '2024-06-15', '2024-12-31'], dtype='datetime64')
print(f"Array de fechas: {fechas_array}")
print(f"Tipo del array: {fechas_array.dtype}")
print()
```

```python

# Diferentes unidades de tiempo
print("🕒 Diferentes precisiones temporales:")
fecha_base = '2024-03-15T14:30:45'
unidades = ['D', 'h', 'm', 's', 'ms', 'us', 'ns']

for unidad in unidades:
    fecha_unidad = np.datetime64(fecha_base, unidad)
    print(f"Unidad '{unidad}': {fecha_unidad}")
```

```python

# Crear rangos de fechas con NumPy
print("\n📅 Rangos de fechas con NumPy:")
rango_fechas = np.arange('2024-01-01', '2024-01-08', dtype='datetime64[D]')
print(f"Semana completa: {rango_fechas}")
```

```python

# Operaciones aritméticas
print("\n🧮 Aritmética con datetime64:")
fecha_inicio = np.datetime64('2024-01-01')
dias_agregar = np.timedelta64(30, 'D')
fecha_final = fecha_inicio + dias_agregar

print(f"Fecha inicio: {fecha_inicio}")
print(f"Agregar: {dias_agregar}")
print(f"Fecha final: {fecha_final}")
```

```python

# Calcular diferencias
fecha1 = np.datetime64('2024-12-31')
fecha2 = np.datetime64('2024-01-01')
diferencia = fecha1 - fecha2
print(f"\nDiferencia entre {fecha1} y {fecha2}: {diferencia}")
```

```python

# Array de timedelta
print("\n⏱️ Array de timedeltas:")
duraciones = np.array([1, 7, 30, 365], dtype='timedelta64[D]')
print(f"Duraciones en días: {duraciones}")
```

```python

# Convertir duraciones a diferentes unidades
print(f"En horas: {duraciones.astype('timedelta64[h]')}")
print(f"En semanas: {duraciones.astype('timedelta64[W]')}")
```

## 8. Datetime en Pandas - Conversión y Creación {#8}

Pandas ofrece herramientas poderosas para trabajar con series temporales y convertir datos de fecha:

```python
# Ejemplo del documento: convertir strings a datetime
data = ['2023-01-20', '2023-04-27', '2023-06-15']
my_series = pd.Series(data)
print("Serie original (strings):")  
display(my_series) # display solo en jupyter
print(f"Tipo de datos: {my_series.dtype}")
print()
```

```python

# Diferentes formatos de entrada
print("🔄 Conversión de diferentes formatos:")
formatos_diversos = pd.Series([
    '2024-03-15',
    '15/03/2024',
    '15-Mar-2024',
    'March 15, 2024',
    '2024-03-15 14:30:00'
])

print("Antes de la conversión:")
display(formatos_diversos)
```

```python

# Pandas es inteligente y detecta automáticamente muchos formatos
fechas_convertidas = pd.to_datetime(formatos_diversos)
print("\nDespués de la conversión:")
print(fechas_convertidas)
```

```python
# Pandas es inteligente y detecta automáticamente muchos formatos
fechas_convertidas = pd.to_datetime(formatos_diversos, format = "mixed")
print("\nDespués de la conversión:")
display(fechas_convertidas)
```

```python

# Para formatos específicos, usar el parámetro format
print("\n🎯 Conversión con formato específico:")
fechas_especificas = pd.Series(['15/03/2024', '20/06/2024', '31/12/2024'])
fechas_formato = pd.to_datetime(fechas_especificas, format='%d/%m/%Y')
display(fechas_formato)
```

```python

# Manejar errores en la conversión
print("\n⚠️ Manejo de errores:")
datos_mixtos = pd.Series(['2024-03-15', 'fecha_invalida', '2024-06-20', 'otro_error'])

# Con errors='coerce', valores inválidos se convierten en NaT (Not a Time)
fechas_con_errores = pd.to_datetime(datos_mixtos, errors='coerce')
print("Con errors='coerce':")
display(fechas_con_errores)
```

```python

# OPCIÓN 4: Función para crear DataFrame multiidioma
def crear_df_multiidioma(fechas_serie):
    """Crea un DataFrame con componentes de fecha en múltiples idiomas"""
    
    df_multi = pd.DataFrame({
        'fecha': fechas_serie,
        'dia_numero': fechas_serie.dt.day,
        'mes_numero': fechas_serie.dt.month,
        'año': fechas_serie.dt.year,
        
        # Inglés (original)
        'dia_en': fechas_serie.dt.day_name(),
        'mes_en': fechas_serie.dt.month_name(),
        
        # Español
        'dia_es': fechas_serie.dt.day_name().map(dias_espanol),
        'mes_es': fechas_serie.dt.month_name().map(meses_espanol),
        
        # Francés
        'dia_fr': fechas_serie.dt.day_name().map(dias_frances),
        'mes_fr': fechas_serie.dt.month_name().map(meses_frances),
        
        # Italiano
        'dia_it': fechas_serie.dt.day_name().map(dias_italiano),
        'mes_it': fechas_serie.dt.month_name().map(meses_italiano)
    })
    
    return df_multi
```

## 10. Operaciones Avanzadas con Fechas en Pandas {#10}

```python
# Aritmética de fechas con Pandas
fecha_base = pd.Timestamp('2024-01-15')
print(f"Fecha base: {fecha_base}")

# Sumar diferentes períodos
print("\n➕ Suma de períodos:")
print(f"+ 30 días: {fecha_base   + pd.Timedelta(days=30)}")
print(f"+ 2 semanas: {fecha_base + pd.Timedelta(weeks=2)}")
print(f"+ 3 meses: {fecha_base   + pd.DateOffset(months=3)}")
print(f"+ 1 año: {fecha_base     + pd.DateOffset(years=1)}")
```

```python

# Crear rangos de fechas con diferentes frecuencias
print("\n📅 Rangos de fechas con diferentes frecuencias:")

frecuencias = {
    'D': 'Diario',
    'W': 'Semanal',
    'M': 'Mensual',
    'Q': 'Trimestral',
    'Y': 'Anual',
    'B': 'Días laborables',
    'H': 'Cada hora',
    '2D': 'Cada 2 días'
}

for freq, descripcion in frecuencias.items():
    if freq in ['H']:  # Para frecuencias altas, limitamos los períodos
        rango = pd.date_range('2024-01-01', periods=5, freq=freq)
    else:
        rango = pd.date_range('2024-01-01', periods=4, freq=freq)
    print(f"{descripcion:15} ({freq:^3}): {rango[0]} ... {rango[-1]} ({len(rango)} períodos)")
```

```python

# Cálculo de diferencias temporales
print("\n⏱️ Cálculo de diferencias temporales:")
fechas_inicio = pd.Series(['2024-01-01', '2024-02-15', '2024-03-30'])
fechas_fin = pd.Series(['2024-01-31', '2024-03-01', '2024-04-15'])

fechas_inicio = pd.to_datetime(fechas_inicio)
fechas_fin = pd.to_datetime(fechas_fin)

diferencias = fechas_fin - fechas_inicio
print("Diferencias:")
for i, diff in enumerate(diferencias):
    print(f"Período {i+1}: {diff} ({diff.days} días)")
```

## 12. Resampleo y Agrupación Temporal {#12}

El resampleo permite cambiar la frecuencia temporal de los datos y realizar agregaciones:

```python
# Usar el DataFrame creado anteriormente para demostrar resampleo
print("📊 Resampleo de datos temporales:")
print(f"Datos originales: {len(datos_ejemplo)} registros diarios")
print()
```

```python

# Resampleo a frecuencias mayores (agregación)
print("🔼 Agregación a frecuencias mayores:")

# Resampleo mensual
mensual = datos_ejemplo.resample('M').agg({
    'ventas': 'sum',
    'clientes': 'mean',
    'temperatura': 'mean'
}).round(2)

print("Resampleo mensual (primeros 6 meses):")
print(mensual.head(6))
print()
```

```python

# Resampleo semanal
semanal = datos_ejemplo.resample('W').agg({
    'ventas': ['sum', 'mean'],
    'clientes': 'mean'
}).round(0)

semanal.columns = ['Ventas_Total', 'Ventas_Promedio', 'Clientes_Promedio']
print("Resampleo semanal (primeras 4 semanas):")
print(semanal.head(4))
print()
```

```python

# Resampleo trimestral
trimestral = datos_ejemplo.resample('Q').agg({
    'ventas': ['sum', 'mean', 'std'],
    'clientes': ['mean', 'max'],
    'temperatura': ['mean', 'min', 'max']
}).round(2)

print("Resampleo trimestral:")
print(trimestral)
print()
```

```python

# Diferentes tipos de agregación
print("🎯 Diferentes métodos de agregación:")

agregaciones_avanzadas = datos_ejemplo.resample('M').agg({
    'ventas': ['sum', 'mean', 'median', 'std', 'min', 'max'],
    'clientes': ['mean', 'sum', 'count']
})

print("Agregaciones múltiples (primeros 3 meses):")
display(agregaciones_avanzadas.head(3))
print()

# Resampleo con función personalizada
print("🔧 Función de agregación personalizada:")

def rango_ventas(series):
    return series.max() - series.min()

custom_resample = datos_ejemplo.resample('M').agg({
    'ventas': [rango_ventas, 'mean'],
    'clientes': 'mean'
})

custom_resample.columns = ['Rango_Ventas', 'Ventas_Promedio', 'Clientes_Promedio']
print("Resampleo con función personalizada (primeros 6 meses):")
print(custom_resample.head(6))
print()
```

```python

# Análisis de tendencias temporales
print("📈 Análisis de tendencias:")

# Calcular medias móviles
datos_ejemplo['ventas_ma_7d'] = datos_ejemplo['ventas'].rolling(window=7).mean()
datos_ejemplo['ventas_ma_30d'] = datos_ejemplo['ventas'].rolling(window=30).mean()

# Mostrar los datos con medias móviles
print("Datos con medias móviles (últimos 10 registros):")
display(datos_ejemplo[['ventas', 'ventas_ma_7d', 'ventas_ma_30d']].tail(10).round(2))
print()
```

```python

# Comparación año a año (simulado)
print("📅 Análisis temporal avanzado:")

# Agregar información temporal adicional
datos_ejemplo['mes'] = datos_ejemplo.index.month
datos_ejemplo['dia_año'] = datos_ejemplo.index.dayofyear
datos_ejemplo['semana_año'] = datos_ejemplo.index.isocalendar().week

# Estadísticas por día de la semana
por_dia_semana = datos_ejemplo.groupby(datos_ejemplo['dia_semana']).agg({
    'ventas': ['mean', 'std'],
    'clientes': 'mean'
}).round(2)

por_dia_semana.columns = ['Ventas_Promedio', 'Ventas_StdDev', 'Clientes_Promedio']
print("Estadísticas por día de la semana:")
display(por_dia_semana)
```

```python

# Resampleo con diferentes anclas (label)
print("\n⚓ Resampleo con diferentes etiquetas:")
resample_left = datos_ejemplo.resample('M', label='left').sum()['ventas'].head(3)
resample_right = datos_ejemplo.resample('M', label='right').sum()['ventas'].head(3)

print("Label='left':")
display(resample_left)
print("\nLabel='right':")
display(resample_right)
```

```python
# Resampleo con diferentes anclas (label) - EXPLICACIÓN DETALLADA
print("\n⚓ Resampleo con diferentes etiquetas:")
print("🔍 ¿Qué significa 'label' en resample?")
print("=" * 50)

# Crear datos de ejemplo para entender mejor
fechas_ejemplo = pd.date_range('2024-01-01', '2024-03-15', freq='D')
ventas_ejemplo = np.random.randint(100, 200, len(fechas_ejemplo))
df_ejemplo = pd.DataFrame({'ventas': ventas_ejemplo}, index=fechas_ejemplo)

print("📊 Datos originales (primeros 10 días):")
display(df_ejemplo.head(10))

print("\n🏷️ LABEL='LEFT' vs LABEL='RIGHT':")
print("=" * 40)

# Resampleo mensual con label='left' (por defecto)
resample_left = df_ejemplo.resample('M', label='left').sum()['ventas']
print("📅 Label='left' (etiqueta al INICIO del período):")
display(resample_left)

# Resampleo mensual con label='right' 
resample_right = df_ejemplo.resample('M', label='right').sum()['ventas']
print("\n📅 Label='right' (etiqueta al FINAL del período):")
display(resample_right)

print("\n💡 EXPLICACIÓN:")
print("─" * 30)
print("• Label='left':  Etiqueta usa la fecha del INICIO del período")
print("  Enero se etiqueta como: 2024-01-01")
print("  Febrero se etiqueta como: 2024-02-01")
print()
print("• Label='right': Etiqueta usa la fecha del FINAL del período") 
print("  Enero se etiqueta como: 2024-01-31")
print("  Febrero se etiqueta como: 2024-02-29")
print()
print("⚠️  LOS DATOS SON LOS MISMOS, solo cambia la ETIQUETA")

# Demostrar que los valores son iguales
print(f"\n🔢 Verificación - valores son iguales:")
print(f"Enero - Left: {resample_left.iloc[0]:,}")
print(f"Enero - Right: {resample_right.iloc[0]:,}")
print(f"¿Son iguales? {resample_left.iloc[0] == resample_right.iloc[0]}")

print("\n🎯 CASOS DE USO:")
print("─" * 20)
print("📈 Label='left' (más común):")
print("  → Para reportes que se refieren al inicio del período")
print("  → 'Ventas de enero' → etiquetado como 2024-01-01")
print()
print("📊 Label='right':")
print("  → Para reportes de cierre de período")
print("  → 'Ventas hasta enero' → etiquetado como 2024-01-31")

# Ejemplo práctico con trimestres
print("\n🗓️ EJEMPLO PRÁCTICO - Trimestres:")
print("─" * 35)
trim_left = df_ejemplo.resample('Q', label='left').sum()['ventas']
trim_right = df_ejemplo.resample('Q', label='right').sum()['ventas']

print("Trimestral con label='left':")
display(trim_left)
print("\nTrimestral con label='right':")
display(trim_right)

print("\n💼 EN RESUMEN:")
print("═" * 25)
print("• Los DATOS son idénticos")
print("• Solo cambia la FECHA de la etiqueta")
print("• 'left' = inicio del período") 
print("• 'right' = final del período")
print("• Por defecto pandas usa 'left'")
print("• Útil para diferentes tipos de reportes")
```

## 📋 Resumen y Puntos Clave

### Funciones Esenciales para Recordar

| **Función** | **Propósito** | **Ejemplo** |
|-------------|---------------|-------------|
| `datetime.strptime()` | String → Datetime | `datetime.strptime("25/11/2022", "%d/%m/%Y")` |
| `datetime.strftime()` | Datetime → String | `dt_object.strftime("%d/%m/%Y")` |
| `pd.to_datetime()` | Conversión inteligente a datetime | `pd.to_datetime(series)` |
| `pd.date_range()` | Crear rangos de fechas | `pd.date_range('2024-01-01', periods=10, freq='D')` |
| `series.dt.accessor` | Extraer componentes | `series.dt.year`, `series.dt.month` |
| `df.resample()` | Agregación temporal | `df.resample('M').mean()` |

### Códigos de Formato Más Utilizados

| **Código** | **Descripción** | **Ejemplo** |
|------------|-----------------|-------------|
| `%Y` | Año (4 dígitos) | 2024 |
| `%m` | Mes (01-12) | 03 |
| `%d` | Día (01-31) | 15 |
| `%H` | Hora 24h (00-23) | 14 |
| `%M` | Minuto (00-59) | 30 |
| `%S` | Segundo (00-59) | 45 |
| `%A` | Día de la semana | Friday |
| `%B` | Nombre del mes | March |

### 💡 Consejos Prácticos

1. **Siempre usar pd.to_datetime()** para conversiones en Pandas - es más robusto que datetime.strptime()
2. **El accesor .dt** es tu mejor amigo para extraer componentes de fechas en Series
3. **Usar resample()** para agregaciones temporales en lugar de groupby cuando sea posible
4. **Manejar zonas horarias** desde el inicio del proyecto para evitar problemas posteriores
5. **Crear índices temporales** facilita enormemente el filtrado y análisis de series temporales

### 🎯 Casos de Uso Comunes

- **Análisis de ventas**: Resampleo mensual/trimestral para identificar tendencias
- **Log analysis**: Filtrado por rangos de fechas para debugging
- **Reportes automáticos**: Extraer períodos específicos usando el accesor dt
- **Coordinación internacional**: Conversión entre zonas horarias para equipos globales
- **Series temporales**: Uso de índices datetime para análisis predictivo
