## 3. Combinar DataFrames con merge()

El método `df.merge()` es una herramienta poderosa para combinar DataFrames basándose en columnas comunes. Veamos diferentes tipos de uniones.

```python
# Crear DataFrames para demostrar merge()
df1 = pd.DataFrame({
    'id': [1, 2, 3, 4],
    'nombre': ['Ana', 'Carlos', 'María', 'Juan'],
    'departamento': ['Ventas', 'IT', 'Marketing', 'Ventas']
})

df2 = pd.DataFrame({
    'id': [1, 2, 3, 5],
    'salario': [50000, 65000, 55000, 60000],
    'bonus': [5000, 8000, 6000, 7000]
})

print("DataFrame 1 (Empleados):")
print(df1)
print("\nDataFrame 2 (Salarios):")
print(df2)
```

```python
# Inner join - solo registros que coinciden en ambos DataFrames
inner_merge = df1.merge(df2, how='inner', on='id')
print("Inner Join:")
print(inner_merge)
```

```python
# Left join - todos los registros del DataFrame izquierdo
left_merge = df1.merge(df2, how='left', on='id')
print("Left Join:")
print(left_merge)
```

```python
# Right join - todos los registros del DataFrame derecho
right_merge = df1.merge(df2, how='right', on='id')
print("Right Join:")
print(right_merge)
```

```python
# Outer join - todos los registros de ambos DataFrames
outer_merge = df1.merge(df2, how='outer', on='id')
print("Outer Join:")
print(outer_merge)
```

## 4. Concatenar DataFrames con concat()

La función `pd.concat()` permite unir DataFrames a lo largo de un eje específico (filas o columnas).

```python
# Crear DataFrames para concatenación
df_2022 = pd.DataFrame({
    'mes': ['Enero', 'Febrero', 'Marzo'],
    'ventas': [12000, 15000, 18000],
    'año': [2022, 2022, 2022]
})

df_2023 = pd.DataFrame({
    'mes': ['Enero', 'Febrero', 'Marzo'],
    'ventas': [15000, 18000, 22000],
    'año': [2023, 2023, 2023]
})

print("Ventas 2022:")
print(df_2022)
print("\nVentas 2023:")
print(df_2023)
```

```python
# Concatenar verticalmente (agregar filas)
concat_vertical = pd.concat([df_2022, df_2023], axis=0, ignore_index=True)
print("Concatenación Vertical (axis=0):")
print(concat_vertical)
```

```python
# Crear DataFrames para concatenación horizontal
df_left = pd.DataFrame({
    'nombre': ['Ana', 'Carlos', 'María'],
    'edad': [25, 30, 28]
})

df_right = pd.DataFrame({
    'salario': [50000, 65000, 55000],
    'departamento': ['Ventas', 'IT', 'Marketing']
})

# Concatenar horizontalmente (agregar columnas)
concat_horizontal = pd.concat([df_left, df_right], axis=1)
print("Concatenación Horizontal (axis=1):")
print(concat_horizontal)
```

## 5. Unir DataFrames con join()

El método `df.join()` es eficiente para combinar DataFrames usando sus índices.

```python
# Preparar DataFrames para join()
df_empleados_info = pd.DataFrame({
    'nombre': ['Ana', 'Carlos', 'María', 'Juan'],
    'departamento': ['Ventas', 'IT', 'Marketing', 'Ventas']
}, index=[101, 102, 103, 104])

df_salarios_info = pd.DataFrame({
    'salario': [50000, 65000, 55000, 48000],
    'bonus': [5000, 8000, 6000, 4800]
}, index=[101, 102, 103, 105])

print("DataFrame Empleados (con índice):")
print(df_empleados_info)
print("\nDataFrame Salarios (con índice):")
print(df_salarios_info)
```

```python
# Join usando índices
join_result = df_empleados_info.join(df_salarios_info, how='inner')
print("Join por índice:")
print(join_result)
```

```python
# Join con diferentes tipos
left_join = df_empleados_info.join(df_salarios_info, how='left')
print("Left Join por índice:")
print(left_join)
```

## 7. Seleccionar Columnas por Tipo de Dato

`df.select_dtypes()` permite filtrar columnas basándose en sus tipos de datos.

```python
# Crear un DataFrame con diferentes tipos de datos
df_mixed = pd.DataFrame({
    'nombre': ['Ana', 'Carlos', 'María'],
    'edad': [25, 30, 28],
    'salario': [50000.5, 65000.0, 55000.75],
    'activo': [True, False, True],
    'fecha_ingreso': pd.to_datetime(['2020-01-15', '2019-03-20', '2021-07-10'])
})

print("DataFrame con tipos mixtos:")
print(df_mixed)
print("\nTipos de datos:")
print(df_mixed.dtypes)
```

```python
# Seleccionar solo columnas numéricas
numericas = df_mixed.select_dtypes(include=['int64', 'float64'])
print("Columnas numéricas:")
print(numericas)
```

```python
# Seleccionar solo columnas de texto
texto = df_mixed.select_dtypes(include=['object'])
print("Columnas de texto:")
print(texto)
```

```python
# Excluir columnas numéricas
no_numericas = df_mixed.select_dtypes(exclude=['int64', 'float64'])
print("Columnas no numéricas:")
print(no_numericas)
```

## 8. Filtrar Datos con Máscaras Booleanas

Las máscaras booleanas son fundamentales para filtrar datos según condiciones específicas.

```python
print("DataFrame de empleados:")
print(df_empleados)

# Filtro simple: empleados con salario mayor a 55000
filtro_salario = df_empleados[df_empleados['salario'] > 55000]
print("\nEmpleados con salario > 55000:")
print(filtro_salario)
```

```python
# Filtro por texto: empleados del departamento de Ventas
filtro_ventas = df_empleados[df_empleados['departamento'] == 'Ventas']
print("Empleados de Ventas:")
print(filtro_ventas)
```

```python
# Filtros múltiples con AND (&)
filtro_multiple = df_empleados[
    (df_empleados['salario'] > 50000) & 
    (df_empleados['años_experiencia'] >= 5)
]
print("Empleados con salario > 50000 Y experiencia >= 5 años:")
print(filtro_multiple)
```

```python
# Filtros múltiples con OR (|)
filtro_or = df_empleados[
    (df_empleados['departamento'] == 'IT') | 
    (df_empleados['salario'] > 60000)
]
print("Empleados de IT O con salario > 60000:")
print(filtro_or)
```

```python
# Filtro con isin() para múltiples valores
departamentos_objetivo = ['IT', 'Marketing']
filtro_isin = df_empleados[df_empleados['departamento'].isin(departamentos_objetivo)]
print("Empleados de IT o Marketing:")
print(filtro_isin)
```

## 9. Ordenar Datos con sort_values()

`df.sort_values()` permite ordenar DataFrames por una o múltiples columnas.

```python
print("DataFrame original de empleados:")
print(df_empleados)

# Ordenar por salario ascendente
ordenado_salario = df_empleados.sort_values(by='salario')
print("\nOrdenado por salario (ascendente):")
print(ordenado_salario)
```

```python
# Ordenar por salario descendente
ordenado_desc = df_empleados.sort_values(by='salario', ascending=False)
print("Ordenado por salario (descendente):")
print(ordenado_desc)
```

```python
# Ordenar por múltiples columnas
ordenado_multiple = df_empleados.sort_values(by=['departamento', 'salario'], ascending=[True, False])
print("Ordenado por departamento (asc) y salario (desc):")
print(ordenado_multiple)
```

## 10. Segmentar Datos con iloc[]

`df.iloc[]` permite seleccionar datos por posición numérica (basado en índices enteros).

```python
print("DataFrame de empleados para iloc:")
print(df_empleados)
print(f"Forma del DataFrame: {df_empleados.shape}")
```

```python
# Seleccionar una fila específica
fila_2 = df_empleados.iloc[2]
print("Fila en posición 2:")
print(fila_2)
```

```python
# Seleccionar un rango de filas
filas_1_3 = df_empleados.iloc[1:4]
print("Filas de la posición 1 a 3:")
print(filas_1_3)
```

```python
# Seleccionar filas y columnas específicas
subset = df_empleados.iloc[0:3, 1:4]
print("Primeras 3 filas, columnas 1-3:")
print(subset)
```

```python
# Seleccionar filas y columnas no consecutivas
no_consecutivas = df_empleados.iloc[[0, 2, 4], [1, 3]]
print("Filas 0, 2, 4 y columnas 1, 3:")
print(no_consecutivas)
```

```python
# Seleccionar última fila y últimas 2 columnas
ultima = df_empleados.iloc[-1, -2:]
print("Última fila, últimas 2 columnas:")
print(ultima)
```

## 11. Segmentar Datos con loc[]

`df.loc[]` permite seleccionar datos por etiquetas, nombres de columnas y condiciones booleanas.

```python
print("DataFrame de empleados para loc:")
print(df_empleados)
```

```python
# Seleccionar filas por índice y columnas por nombre
subset_loc = df_empleados.loc[0:2, ['nombre', 'departamento']]
print("Filas 0-2, columnas específicas:")
print(subset_loc)
```

```python
# Seleccionar todas las filas y columnas específicas
columnas_especificas = df_empleados.loc[:, ['nombre', 'salario']]
print("Todas las filas, columnas específicas:")
print(columnas_especificas)
```

```python
# Usar condiciones booleanas con loc
condicion_loc = df_empleados.loc[df_empleados['salario'] > 55000, ['nombre', 'salario', 'departamento']]
print("Empleados con salario > 55000:")
print(condicion_loc)
```

```python
# Combinar condiciones múltiples con loc
condicion_multiple = df_empleados.loc[
    (df_empleados['departamento'] == 'IT') & (df_empleados['años_experiencia'] > 5),
    ['nombre', 'salario', 'años_experiencia']
]
print("Empleados de IT con experiencia > 5 años:")
print(condicion_multiple)
```

## 12. Ejercicios Prácticos Combinados

Ahora practicaremos combinando múltiples técnicas para resolver problemas reales de estructuración de datos.

```python
# Crear datos más complejos para ejercicios
df_ventas_detalle = pd.DataFrame({
    'id_venta': [1, 2, 3, 4, 5, 6],
    'id_empleado': [1, 2, 3, 1, 4, 5],
    'producto': ['Laptop', 'Mouse', 'Teclado', 'Monitor', 'Laptop', 'Mouse'],
    'cantidad': [2, 10, 5, 1, 1, 15],
    'precio_unitario': [800.0, 25.0, 50.0, 300.0, 800.0, 25.0],
    'fecha': pd.to_datetime(['2023-01-15', '2023-01-16', '2023-01-17', '2023-01-18', '2023-01-19', '2023-01-20'])
})

df_productos = pd.DataFrame({
    'producto': ['Laptop', 'Mouse', 'Teclado', 'Monitor'],
    'categoria': ['Computadoras', 'Accesorios', 'Accesorios', 'Computadoras'],
    'costo': [600.0, 15.0, 30.0, 200.0]
})

print("Ventas detalladas:")
print(df_ventas_detalle)
print("\nInformación de productos:")
print(df_productos)
```

### Ejercicio 1: Análisis Completo de Ventas

Combinaremos múltiples DataFrames y realizaremos análisis usando todas las técnicas aprendidas.

```python
# Paso 1: Combinar ventas con información de empleados
ventas_empleados = df_ventas_detalle.merge(df_empleados, on='id_empleado', how='left')
print("Ventas con información de empleados:")
print(ventas_empleados)
```

```python
# Paso 2: Agregar información de productos
ventas_completo = ventas_empleados.merge(df_productos, on='producto', how='left')
print("Información completa de ventas:")
print(ventas_completo)
```

```python
# Paso 3: Calcular métricas importantes
ventas_completo['total_venta'] = ventas_completo['cantidad'] * ventas_completo['precio_unitario']
ventas_completo['ganancia'] = ventas_completo['total_venta'] - (ventas_completo['cantidad'] * ventas_completo['costo'])

print("Ventas con métricas calculadas:")
print(ventas_completo[['id_venta', 'nombre', 'producto', 'total_venta', 'ganancia']])
```

### Ejercicio 2: Análisis por Segmentos

```python
# Filtrar ventas de computadoras con alta ganancia
ventas_computadoras = ventas_completo[
    (ventas_completo['categoria'] == 'Computadoras') & 
    (ventas_completo['ganancia'] > 150)
]
print("Ventas de computadoras con ganancia > 150:")
print(ventas_computadoras[['nombre', 'producto', 'total_venta', 'ganancia']])
```

```python
# Análisis por departamento
ventas_por_depto = ventas_completo.groupby('departamento').agg({
    'total_venta': 'sum',
    'ganancia': 'sum',
    'cantidad': 'sum'
}).round(2)

print("Resumen por departamento:")
print(ventas_por_depto)
```

```python
# Top 3 ventas más rentables
top_ventas = ventas_completo.sort_values('ganancia', ascending=False).iloc[:3]
print("Top 3 ventas más rentables:")
print(top_ventas[['nombre', 'producto', 'total_venta', 'ganancia']])
```

### Ejercicio 3: Segmentación Avanzada

```python
# Seleccionar solo empleados de IT con sus ventas
ventas_it = ventas_completo.loc[
    ventas_completo['departamento'] == 'IT',
    ['nombre', 'producto', 'cantidad', 'total_venta', 'fecha']
]
print("Ventas del departamento IT:")
print(ventas_it)
```

```python
# Crear resumen ejecutivo combinando todas las técnicas
resumen_ejecutivo = ventas_completo.groupby(['departamento', 'categoria']).agg({
    'total_venta': ['sum', 'mean'],
    'ganancia': ['sum', 'mean'],
    'id_venta': 'count'
}).round(2)

resumen_ejecutivo.columns = ['Total_Ventas', 'Promedio_Ventas', 'Total_Ganancia', 'Promedio_Ganancia', 'Num_Transacciones']
print("Resumen ejecutivo por departamento y categoría:")
print(resumen_ejecutivo)
```
