## 1. ¿Qué son los entornos virtuales?

Un **entorno virtual** es una carpeta aislada que contiene su propia instalación de Python y sus paquetes, separada del sistema global.

### ¿Por qué existen?

Imaginá que tenés dos proyectos:
- **Proyecto A** necesita `pandas 1.5`
- **Proyecto B** necesita `pandas 2.0`

Si instalás ambas versiones globalmente, una sobreescribe a la otra. Con un entorno virtual, cada proyecto tiene sus propias dependencias sin conflictos.

### Ventajas principales

| Ventaja | Descripción |
|---------|-------------|
| **Aislamiento** | Cada proyecto tiene sus paquetes sin interferir con otros |
| **Reproducibilidad** | "Funciona en mi máquina" se acabó: el entorno se puede replicar |
| **Flexibilidad** | Probar versiones nuevas sin romper lo existente |
| **Limpieza** | Si algo sale mal, borrás el entorno y creás uno nuevo |

## 2. Crear entorno con `venv`

`venv` viene incluido con Python, no necesitás instalar nada extra.

### Linux / macOS

```bash
# Crear entorno
python3 -m venv mi_entorno

# Activar entorno
source mi_entorno/bin/activate

```

### Windows (PowerShell)

```powershell
# Crear entorno
python -m venv mi_entorno

# Activar entorno
.\mi_entorno\Scripts\Activate.ps1

# Desactivar entorno
deactivate
```

### Windows (CMD)

```cmd
# Crear entorno
python -m venv mi_entorno

# Activar entorno
.\mi_entorno\Scripts\activate.bat

# Desactivar entorno
deactivate
```

> **Tip:** Cuando el entorno está activado, verás `(mi_entorno)` al inicio de la línea de comandos.

## 3. `uv`: gestión ultrarrápida de entornos

[`uv`](https://docs.astral.sh/uv/) es una herramienta escrita en Rust que reemplaza a `pip` y `venv` con velocidades de instalación **10-100x más rápidas**.

### Instalación

```bash
# Windows (PowerShell)
pip install uv

# Linux / macOS
pip install uv
# o con curl
curl -LsSf https://astral.sh/uv/install.sh | sh
```

### Crear un entorno con uv

```bash
# Crear entorno virtual (más rápido que python -m venv)
uv venv

# Crear entorno con nombre específico
uv venv mi_entorno

# Crear con una versión de Python específica
uv venv --python 3.12
```

### Instalar paquetes con uv

```bash
# Activar el entorno primero (misma sintaxis que venv)
source mi_entorno/bin/activate      # Linux/macOS
.\mi_entorno\Scripts\Activate.ps1  # Windows

# Instalar un paquete (ultrarrápido)
uv pip install pandas

# Instalar varios paquetes
uv pip install pandas numpy matplotlib

# Instalar desde un archivo requirements.txt
uv pip install -r requirements.txt

# Guardar dependencias del entorno actual
uv pip freeze > requirements.txt
```

### Comparación de velocidad

| Herramienta | Tiempo aprox. (pandas + numpy) |
|-------------|-------------------------------|
| `pip install` | ~30-60 segundos |
| `uv pip install` | ~2-5 segundos |

## 5. Hacer un `.py` ejecutable en Linux

En Linux (y macOS) podés ejecutar un script Python directamente como si fuera un programa, sin escribir `python`.

### Paso 1: Agregar shebang

Al **principio** del archivo `.py`, agregá esta línea:

```python
#!/usr/bin/env python3
```

Esto le dice al sistema: "usá el Python que encuentres en el PATH".

### Paso 2: Dar permisos de ejecución

```bash
chmod +x script.py
```

### Paso 3: Ejecutar directamente

```bash
./script.py
# o con argumentos
./script.py --name "Juan"
```

### Ejemplo completo

```bash
# 1. Navegá a la carpeta
cd Automatizacion/Automatizacion/Modulo_1

# 2. Dá permisos
chmod +x hello.py

# 3. Ejecutá
./hello.py --name "Juan"
```

> **Nota:** El shebang `#!/usr/bin/env python3` es **recomendado** porque es portable: funciona sin importar dónde esté instalado Python.


## 7. Scripts de shell: `.sh`, `.bat`, `.ps1`

### Script Bash (`.sh`) — Linux / macOS

```bash
#!/bin/bash
# archivo: ejecutar_formateo.sh

echo "=== Formateando archivos de texto ==="

# Activar entorno virtual
source ../mi_entorno/bin/activate

# Ejecutar el script Python
python formatear_txt.py "$@"

echo "Listo!"
```

**Cómo ejecutarlo:**
```bash
chmod +x ejecutar_formateo.sh
./ejecutar_formateo.sh intro.txt -w 80
```

### Script CMD (`.bat`) — Windows

```bat
@echo off
REM archivo: ejecutar_formateo.bat

echo === Formateando archivos de texto ===

REM Activar entorno virtual
call ..\mi_entorno\Scripts\activate.bat

REM Ejecutar el script Python
python formatear_txt.py %*

echo Listo!
```

**Cómo ejecutarlo:**
```cmd
ejecutar_formateo.bat intro.txt -w 80
```

### Script PowerShell (`.ps1`) — Windows

```powershell
# archivo: ejecutar_formateo.ps1

Write-Host "=== Formateando archivos de texto ===" -ForegroundColor Cyan

# Activar entorno virtual
& "$PSScriptRoot\..\mi_entorno\Scripts\Activate.ps1"

# Ejecutar el script Python
python "$PSScriptRoot\formatear_txt.py" @args

Write-Host "Listo!" -ForegroundColor Green
```

**Cómo ejecutarlo:**
```powershell
.\ejecutar_formateo.ps1 intro.txt -w 80
```

### Diferencias clave

| Característica | `.sh` | `.bat` | `.ps1` |
|---------------|-------|--------|--------|
| Sistema | Linux/macOS | Windows | Windows |
| Shell | bash | CMD | PowerShell |
| Activar entorno | `source activate` | `call activate.bat` | `& Activate.ps1` |
| Argumentos | `$@` | `%*` | `@args` |

## 8. Manejo de argumentos: `sys.argv` vs `argparse`

### `sys.argv`: acceso directo

`sys.argv` es una lista simple con los argumentos de la línea de comandos.

````python
%%writefile sumar.py
# Ejemplo con sys.argv
# Ejecutar desde terminal: python script.py hola mundo

import sys

print(f"Nombre del script: {sys.argv[0]}")
print(f"Argumentos: {sys.argv[1:]}")
print(f"Cantidad de argumentos: {len(sys.argv) - 1}")

# Ejemplo práctico: sumar números pasados por CLI
# Ejecutar: python script.py 10 20 30
if len(sys.argv) > 1:
    numeros = [int(x) for x in sys.argv[1:]]
    print(f"Suma de {numeros} = {sum(numeros)}")
else:
    print("no pasaste nada")
````

### `argparse`: manejo robusto

`argparse` es la librería estándar para crear CLIs profesionales con:
- Ayuda automática (`--help`)
- Validación de tipos
- Argumentos opcionales y posicionales
- Valores por defecto

````python
%%writefile primer_cli.py
# Ejemplo con argparse
# Guardá este código como mi_cli.py y ejecutá: python mi_cli.py --nombre Juan --edad 25

import argparse

parser = argparse.ArgumentParser(description="Mi primera CLI con argparse")

# Argumento obligatorio
parser.add_argument("nombre", help="Tu nombre")

parser.add_argument("apellido", help="tu apellido")

# Argumento opcional con valor por defecto
parser.add_argument("--edad", type = int, default = 18, help = "Tu edad (default: 18)")

# Argumento con opciones predefinidas
parser.add_argument("--color", choices = ["rojo", "azul", "verde"], help="Color favorito")

args = parser.parse_args()

print(f"Hola {args.nombre} {args.apellido}, tenés {args.edad} años.")
if args.color:
    print(f"Tu color favorito es {args.color}.")
````

## 10. Ejercicio práctico: `formatear_txt.py`

Vamos a analizar el script `formatear_txt.py` que tenemos en la carpeta del módulo 1.

### Versión básica: CLI con argparse

Este script toma archivos `.txt` y reformatea el texto ajustándolo a un ancho de columna específico.

````python
%%writefile formatear.py
# formatear_txt.py — Versión básica con argparse
# Ejecutar desde terminal:
#   python formatear_txt.py intro.txt -w 80
#   python formatear_txt.py  (procesa todos los .txt del directorio)

import argparse
import sys
from pathlib import Path
from textwrap import TextWrapper


def formatear_archivo(ruta: Path, ancho: int) -> Path:
    """Lee un archivo .txt y crea una versión formateada con el ancho indicado."""
    wrapper = TextWrapper(
        width=ancho,
        break_long_words=False,
        break_on_hyphens=False,
        replace_whitespace=True,
    )

    lineas_salida = []
    for linea in ruta.read_text(encoding="utf-8").splitlines():
        if not linea.strip():
            lineas_salida.append("")
        else:
            lineas_salida.extend(wrapper.wrap(linea))

    ruta_salida = ruta.with_name(f"{ruta.stem}_{ancho}.txt")
    ruta_salida.write_text("\n".join(lineas_salida) + "\n", encoding="utf-8")
    return ruta_salida


def main():
    parser = argparse.ArgumentParser(
        description="Reformatea archivos TXT envolviendo el texto con textwrap."
    )
    parser.add_argument(
        "archivos", nargs="*",
        help="Archivos .txt a procesar (por defecto, todos los del directorio actual)"
    )
    parser.add_argument(
        "-w", "--ancho", type=int, default=100,
        help="Ancho máximo de columna (default: 100)"
    )
    args = parser.parse_args()

    if args.archivos:
        rutas = [Path(a) for a in args.archivos]
    else:
        rutas = sorted(Path(".").glob("*.txt"))

    if not rutas:
        print("No se encontraron archivos .txt para procesar.")
        sys.exit(1)

    for ruta in rutas:
        if not ruta.is_file():
            print(f"AVISO: {ruta} no existe, se omite.")
            continue
        salida = formatear_archivo(ruta, args.ancho)
        print(f"OK: {ruta} -> {salida}")


if __name__ == "__main__":
    main()
````

### Versión mejorada: CLI interactiva con `rich`

Ahora vamos a mejorar el mismo script usando `rich` para:
- Mostrar los resultados con **colores**
- Crear una **tabla** con los archivos procesados
- Mostrar una **barra de progreso** durante el procesamiento

````python
# formatear_txt_rich.py — Versión mejorada con rich
# Ejecutar desde terminal:
#   python formatear_txt_rich.py intro.txt -w 80
#   python formatear_txt_rich.py  (procesa todos los .txt del directorio)

import argparse
import sys
from pathlib import Path
from textwrap import TextWrapper

from rich.console import Console
from rich.table import Table
from rich.panel import Panel
from rich.progress import track

console = Console()


def formatear_archivo(ruta: Path, ancho: int) -> Path:
    """Lee un archivo .txt y crea una versión formateada con el ancho indicado."""
    wrapper = TextWrapper(
        width=ancho,
        break_long_words=False,
        break_on_hyphens=False,
        replace_whitespace=True,
    )

    lineas_salida = []
    for linea in ruta.read_text(encoding="utf-8").splitlines():
        if not linea.strip():
            lineas_salida.append("")
        else:
            lineas_salida.extend(wrapper.wrap(linea))

    ruta_salida = ruta.with_name(f"{ruta.stem}_{ancho}.txt")
    ruta_salida.write_text("\n".join(lineas_salida) + "\n", encoding="utf-8")
    return ruta_salida


def main():
    parser = argparse.ArgumentParser(
        description="[bold cyan]Reformatea archivos TXT con estilo[/bold cyan]"
    )
    parser.add_argument(
        "archivos", nargs="*",
        help="Archivos .txt a procesar"
    )
    parser.add_argument(
        "-w", "--ancho", type=int, default=100,
        help="Ancho máximo de columna (default: 100)"
    )
    args = parser.parse_args()

    console.print(Panel(
        f"[bold]Ancho de columna:[/bold] {args.ancho}",
        title="[bold green]Formateador de texto[/bold green]",
        border_style="green"
    ))

    if args.archivos:
        rutas = [Path(a) for a in args.archivos]
    else:
        rutas = sorted(Path(".").glob("*.txt"))

    if not rutas:
        console.print("[red]No se encontraron archivos .txt para procesar.[/red]")
        sys.exit(1)

    # Procesar con barra de progreso
    resultados = []
    for ruta in track(rutas, description="[cyan]Procesando..."):
        if not ruta.is_file():
            console.print(f"[yellow]AVISO:[/yellow] {ruta} no existe, se omite.")
            continue
        try:
            salida = formatear_archivo(ruta, args.ancho)
            lineas_orig = len(ruta.read_text(encoding="utf-8").splitlines())
            lineas_nuev = len(salida.read_text(encoding="utf-8").splitlines())
            resultados.append((str(ruta), str(salida), lineas_orig, lineas_nuev))
            console.print(f"  [green]OK[/green]: {ruta} -> {salida}")
        except Exception as e:
            console.print(f"  [red]ERROR[/red]: {ruta} — {e}")

    # Mostrar tabla resumen
    if resultados:
        tabla = Table(title="Archivos procesados")
        tabla.add_column("Original", style="cyan")
        tabla.add_column("Salida", style="green")
        tabla.add_column("Líneas orig.", justify="right")
        tabla.add_column("Líneas nuevas", justify="right")

        for orig, sal, lo, ln in resultados:
            tabla.add_row(orig, sal, str(lo), str(ln))

        console.print()
        console.print(tabla)
        console.print(f"\n[bold green]¡Listo! {len(resultados)} archivo(s) procesado(s).[/bold green]")


if __name__ == "__main__":
    main()
````
